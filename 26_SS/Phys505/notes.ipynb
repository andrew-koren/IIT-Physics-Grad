Green's theorem. Going beyond $\Phi(\vec x)$

$$
\int_V \vec \nabla \cdot \vec A d^3 x = \oint_S \vec A \cdot \hat n da
$$

Let's consider a particular $\vec A = \phi \vec \nabla \psi$, then we have

$$
      \int_V \left( \vec \nabla \phi \cdot \vec \nabla \psi + \phi \nabla^2 \psi \right) d^3 x 
    = \oint_S \phi {\partial \psi \over \partial n} da
\\    \vdots
\\    \int_V \left(\phi \nabla^2 \psi - \psi \nabla^2 \phi \right) d^3 x 
    = \oint_S \left( \phi {\partial \psi \over \partial n} - \psi {\partial \phi \over \partial n}\right) da
$$

For a particular case $\phi=\Phi$ and $\psi = {1 \over R}$ 
(remember $\nabla^2 \psi = -4\pi {\delta \vec x - \vec x'}$),
we have

$$
\def\d#1#2{\frac{\partial #1}{\partial #2}}
    \int_V \left[ 
        -4\pi \Phi(\vec x') \delta(\vec x - \vec x')+{1\over R} {\rho (\vec x')\over \epsilon_0} 
    \right]d^3x'
    =
      \oint_S \left[ 
         \Phi(\vec x') { 1 \over \partial n'} \left(1 \over R \right) \d{\Phi}{n'}
        -{1 \over R} \d{\Phi}{n'}
      \right] da'

\\    \Phi(\vec x) %green's theorem
    = {1 \over 4 \pi \epsilon_0} \int_V 
      {\rho(\vec x') \over R} d^3 x' 
    + {1 \over 4\pi } \oint_S \left[ 
        {1 \over R} {\partial \Phi \over \partial n'} 
      - \Phi(\vec x ') {\partial \over \partial n'} \left(1 \over R \right) 
      \right] da'
$$

Uniqueness of poisson equation $\nabla^2 \Phi = {\rho \over \epsilon_0}$ in slide ~23

Free Space Green Function

$$
      \Phi_0 
    = {1 \over 4\pi \epsilon_0} {1 \over |\vec x - \vec x'|} \equiv G_0(\vec x, \vec x')
\\    \Phi(\vec x) = \int_V \rho(\vec x')G_0(\vec x, \vec x') d^3 x' % convolution
$$
$\Phi$ is found via convolution.

Now we can superposition things $G(\vec x, \vec x') = G_0(\vec x,\vec x') + F(\vec x, \vec x')$

Whiteboard Statement:

Given an arbitrary $\rho(\vec x' \in V)$ and $\Phi(\vec x' \in S)$, we can compute $\Phi(\vec x \in V)$ if we know $G_D(\vec x, \vec x')$. For Neumann, it's $G_N$ if we have an arbitrary $\partial_{n'} \Phi(\vec x' \in S)$

Neumann boundary conditions 
$$
\def\d#1#2{\frac{\partial #1}{\partial #2}}
    \d{G_n}{n'} = -{4\pi \over s}\text{, for } \vec x' \in S
\\  s = \oint_S da \text{, area of the surface}
$$
This yields
$$
\Phi(\vec x) = \langle \Phi \rangle_S + {1 \over 4\pi \epsilon_0} \int_V \rho(\vec x') G_N(\vec x, \vec x') d^3 x' + {1 \over 4\pi} \oint_S G_N \d{\Phi}{n'}da'
$$

This is kinda a generalization of images / gauss's law / equipotential problems, none of the crap really goes away. In fact, this is the main point of the entire course.
