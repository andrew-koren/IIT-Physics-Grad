Use the easiest surface to describe the electric field of the charge distribution:
$$
\oint_S \vec E \cdot \hat n da = {1 \over \epsilon_0} \int_V \rho(\vec x) d^3 x = {Q_{in}\over \epsilon_0}
$$
Works for gravity too!

(Not) magical math identity: Divergence theorem & FTC & work

$$
    \oint_S \vec A \cdot \hat n da = \int_V \vec \nabla \cdot \vec A d^3 x
\\  \int_{\vec A}^{\vec B} \vec \nabla f \cdot d \vec l = \int_{\vec A}^{\vec B} df = f(\vec B) - f(\vec A)
\\  W_{\vec A \rightarrow \vec B} = - \int_{\vec A}^{\vec B} \vec F \cdot d\vec l
$$

Useful identity:

$$
\vec \nabla \left( {1 \over \left|\vec x - \vec x' \right|}\right) = -{\vec x - \vec x' \over \left|\vec x - \vec x' \right|^3}
$$
Use backwards to create a useful $\vec \nabla$ when working with $\vec E$ & divergence theorem. This trick is use to derive
$$
    \Phi(\vec x) = k \int{\rho(\vec x ' ) \over \left|\vec x - \vec x'\right|} d^3 x'
% \\  W = -q\int_{\vec A}^{\vec B} \vec E \cdot d \vec l = q \int_{\vec A}^{\vec B} \vec \nabla 
$$
Also surface charge density via divergence from charge
$$
    \Delta E_{\perp} = {\sigma \over \epsilon_0}
\\  \Delta E_{\parallel} = 0
$$
Note: surface charge keeps potential continuous (check the slides), but a surface dipole layer could break it.
