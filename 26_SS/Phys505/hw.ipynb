Gauss's law
$$
\begin{gathered}
\Phi_E   = {Q \over \epsilon_0}
         = \int_S \vec E \cdot d\vec A
% \\ \vec \nabla \cdot \vec E  = {\rho \over \epsilon_0}
\end{gathered}
$$
Electric field to voltage
$$
E = -\nabla V \Rightarrow \Delta V = - \int_{x_1}^{x_2} \vec E \cdot d \vec l
$$

Using gaussian shells we can easily compute the electric field.

(a) With $A \gg d$ we assume a uniform field between plates. Using two rectangular prism surfaces each enclosing and cutting through a plate with shape $A\times x$ and $A\times (d-x)$, we'll set plate 1 with charge $Q$ at $x=0$ and plate 2 with charge $-Q$ at $x=d$, the field at a point $x$ between plates is found via superposition as

$$
\begin{gathered}
\begin{gathered}
      {Q \over \epsilon_0} 
    = \int_{S_1} \vec E_1 \cdot d\vec A
    = E_1 A,
&
    \vec E_1 = {Q \over \epsilon_0 A} \hat x,
\\   -{Q \over \epsilon_0} 
    = \int_{S_2} \vec E_2 \cdot d\vec A
    = -E_2 A,
& 
    \vec E_2 = {Q \over \epsilon_0 A} \hat x,
\end{gathered}
\\  \vec E = \vec E_1 + \vec E_2 = {2 Q \over \epsilon_0 A} \hat x
\\  {Q \over \Delta V} = {2Q \over Ed} = {\epsilon_0}{A \over d}


\end{gathered}
$$

(b) The field will be uniform, so we can ignore the outer shell. For a spherical gaussian surface $a<r<b$ we'll have

$$
\begin{gathered}
{Q \over \epsilon_0} = \int_S E(r) \hat r \cdot d \vec A = E\times 4 \pi r^2,
\\  E(r) = {Q \over 4\pi r^2}, \quad \vec E = {Q \over 4 \pi r^2} \hat r,
\\  \Delta V = -\int_a^b E(r) dr = \left.{Q \over 4\pi r} \right|_a^b
\\  {Q \over \Delta V} = 4\pi\left({ab \over b-a}\right)
\end{gathered}
$$

(c) same deal as the previous, using $L \gg b >a$ we can assume a uniform field between plates
$$
\begin{gathered}
{Q \over \epsilon_0} = \int_{S_1} E(r)\hat r \cdot d\vec A = E \times 2\pi r l,
\\  E(r) = {Q \over 2 \pi r l \epsilon_0 }, \quad \vec E(r) = E(r) \hat r,
\\  \Delta V = - \int_b^a E(r)dr = -{Q \over 2 \pi l \epsilon_0 }\left.\ln(r)\right|_b^a
\\  {Q \over \Delta V} = 2\pi l \epsilon_0 \ln\left(b\over a\right)
\end{gathered}
$$

(d) ${C \over l} = 2 \pi \epsilon_0 \ln \left(b/ a\right)$, so $b = a\exp\left({C \over l}\big/2\pi \epsilon_0\right)$

```python
import numpy as np
from scipy.constants import epsilon_0


F_m1 = 3e-11
F_m2 = 3e-12
a = 1e-3
print(f'b = {a * np.exp(F_m1/(2*np.pi*epsilon_0)) * 1000:.3f} mm')
print(f'b = {a * np.exp(F_m2/(2*np.pi*epsilon_0)) * 1000:.3f} mm')
```
