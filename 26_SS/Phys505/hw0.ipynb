## 0.2

Using einstein summation $a_i b_i = \sum_{i=1}^{3} a_i b_i$ and $\frac{\partial}{\partial x_i} = \partial_i$

1. $\vec \nabla r = \hat n$

I think we need a new definition. $r^2 = x_i x_i$, so $r = \sqrt{x_j x_j}$

$$
\begin{align*}
      \left[\vec \nabla r  \right]_i 
   &= \partial_i \sqrt{x_j x_j}
    = {1 \over 2 \sqrt{x_j x_j}} \partial_i \left( x_j x_j \right)
\\ &= {1 \over 2r} \left((\partial_i x_j) x_j + x_j (\partial_i x_j)\right)
\\ &= {1 \over 2r} \left(\delta_{ij} x_j + x_j \delta_{ij}          \right)
    = {2 x_i \over 2r}
\\ &= {x_i \over r} = \left[\hat n\right]_i
\end{align*}
$$


2. $\vec \nabla \cdot \vec x = 3$

$$
[\vec \nabla \cdot \vec x]_i = \partial_i x_i = \delta_{ii} = 1
$$
repeat for all three dimensions

3. $\vec \nabla \times \vec x = 0$

$$
\left[\vec \nabla \times \vec x \right]_i = \epsilon_{ijk} \partial_j x_k = \epsilon_{ijk} \delta_{jk} = 0
$$
Since $\epsilon$ is zero for $j=k$ and $\delta$ is zero for $j\neq k$, this sums to zero.
$$
\epsilon_{ijk} \delta_{jk} = \epsilon_{ijj} = 0
$$

4. $\nabla^2 r = {2 \over r}$

Fortunately the scalar laplacian is simple. Using some work from identitiy 1., we have

$$
      \nabla^2 r 
    = \partial_i \partial_i r 
    = \partial_i {x_i \over r}
    = {r \partial_i x_i - x_i \partial_i r \over r^2}
\\  = {\delta_{ii} r - x_i { x_i \over r} \over r^2}
    = {3 r - r \over r^2}
    = {2 \over r}
$$

5. Combine result from 4. with result from 1.

6. Divie result from 3. by $r$

7. $\nabla^2 r^2 = 6$

$$
      \partial_i \partial_i x_j x_j 
    = \partial_i \left((\partial_i x_j) x_j + x_j (\partial_i x_j)\right)
\\  = \partial_i \left(2 \delta_{ij} x_j\right)
    = 2 \partial_i x_i 
    = 2 \delta_{ii}
    = 6
$$

8. $\vec \nabla f(r) = \hat n \partial_r f$

This is just chain rule. 

$$
      \left[ \vec \nabla f(r) \right]_i
    = \partial_i (f(r))
    = (\partial_i r)(\partial_r f) 
\\[0.25 em]
    = {x_i \over r} (\partial_r f) 
    = \left[ \hat n \right]_i \partial_r f 

$$

9. $\vec \nabla \cdot \left[ \hat n f \right] = {2 \over r} f + \partial_r f$

$$
      \vec \nabla \cdot \left[ \hat n  f \right]
    = \partial_i \left( {x_i \over r} ( f)  \right)
    = \underbrace{\left( \partial_i  {x_i \over r} \right)}_{4.}  f
    + {x_i \over r}  \underbrace{\left( \partial_i f \right)}_{8.}
$$
apply result from 4. and 8.
$$
    = {2 \over r} f + {x_i x_i \over r^2} \partial_r f
    = {2 \over r} f + \partial_r f
$$

10. $\vec \nabla \times \left[ \hat n f(r) \right] = \vec 0 $

$$
      \left[ \vec \nabla \times \left[ \hat n f(r) \right] \right]_i
    = \epsilon_{ijk} \partial_j {x_k \over r} f
    = \epsilon_{ijk} \delta_{jk} {f \over r} = 0
$$
as per logic from 3.
