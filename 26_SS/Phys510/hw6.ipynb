```python
from IPython.display import display, Latex
from sympy import Matrix, I, sqrt, zeros, eye, simplify, symbols, Eq, MatMul, expand
from sympy.tensor.tensor import tensor_indices, tensor_heads, TensorHead, TensorSymmetry
from sympy.physics.hep.gamma_matrices import GammaMatrix as G, simplify_gamma_expression, gamma_trace, LorentzIndex
from sympy import latex

I2 = eye(2)
Z2 = zeros(2)

s1 = Matrix([[0, 1], [1, 0]])
s2 = Matrix([[0, -I], [I, 0]])
s3 = Matrix([[1, 0], [0, -1]])

def block(A, B, C, D):
    top = A.row_join(B)
    bot = C.row_join(D)
    return top.col_join(bot)

gW0 = block(Z2, I2, I2, Z2)
gW1 = block(Z2, -s1, s1, Z2)
gW2 = block(Z2, -s2, s2, Z2)
gW3 = block(Z2, -s3, s3, Z2)
g5W = block(I2, Z2, Z2, -I2)  # i*g0*g1*g2*g3 in Weyl

expr = Eq(g5W, MatMul(I,gW0,gW1,gW2,gW3,evaluate=False), evaluate=False)
display(Latex(r'We define $\gamma^5$ as $' + latex(expr) + r'$'))

eigs0 = gW0.eigenvects()
display(Latex(r'$\text{Eigenvalues and eigenvectors of } \gamma^0 \text{ (Weyl rep):}$'))

for eigenval, mult, vecs in eigs0:
    vec_text = r',\quad '.join(latex(v.T) for v in vecs)

    display(Latex(
        fr'$\lambda = {latex(eigenval)},'
        fr'\quad \text{{mult}} = {mult},'
        fr'\quad \text{{eigenvectors: }} {vec_text}$'
    ))

# Transformation matrix U (Weyl -> Dirac)
U = (1/sqrt(2)) * block(I2, I2, I2, -I2)
Uinv = U.inv()
print('Unitary Transformation')
display(Eq(symbols('U'), U, evaluate=False))
display(Eq(symbols('U^{-1}'), Uinv, evaluate=False))

gD0 = simplify(U * gW0 * Uinv)
gD1 = simplify(U * gW1 * Uinv)
gD2 = simplify(U * gW2 * Uinv)
gD3 = simplify(U * gW3 * Uinv)
g5D = simplify(U * g5W * Uinv)

gammas = [gD0, gD1, gD2, gD3, g5D]

display(Latex(r'$\text{Dirac representation (from Weyl via } U\text{):}$'))
for i, gam in enumerate(gammas):
    label = '5' if i == 4 else str(i)
    display(Latex(fr'$\gamma^{label}_D = {latex(gam)}$'))

```
