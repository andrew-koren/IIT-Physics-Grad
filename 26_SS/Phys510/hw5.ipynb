We are studying free complex scalar fields $\phi = \phi_1 + i\phi_2$ which exhibit $U(1)$ global phase symmetry, i.e. we can multiply by an arbitrary phase factor $e^{i\theta}$. We'll start from classical mechanics and work our way to quantization. The Lagrangian for the system is

$$
\def\l{\left}\def\r{\right}
\def\c#1#2{\l[#1,#2\r]}

\def\d#1#2{\frac{\partial #1}{\partial #2}}

\def\L{\mathcal L}
$$

$$
\L = {1 \over 2} \p_\mu \phi^* \p^\mu \phi - {1 \over 2} m^2 \phi^* \phi.
$$
We would like to derive the Euler-Lagrange equations via least action principle. The action is $S = \int \L d^4 x$. This should collapse to Klein-Gordon-Fock for a special case

$$
\sq \phi + m^2 \phi = 0.
$$

Note we're using the minkowski metric from class with diagonal $(1,-1,-1,-1)$, and we use shorthand $\p_\mu = {\p \over \p x^\mu}$. Also note $g^{\mu \nu} = g_{\mu \nu}$.

$$
\begin{align*}
        \delta S 
   &= \int d^4 x\ \delta\left(
          {1 \over 2} \p_\mu \phi^* \p^\mu \phi 
        - {1 \over 2} m^2 \phi^* \phi 
      \right),
\\ &= \int d^4 x\ \left[ \delta\left(
          {1 \over 2} \p_\mu \phi^* \p^\mu \phi
      \right) 
    - \delta \left(
          {1 \over 2} m^2 \phi^* \phi
        \right) \right],
\\ &= {1\over 2}\int d^4 x\ \left[    
          \p_\mu \delta \phi^* \p^\mu \phi
        + \p_\mu \phi^* \p^\mu \delta \phi
             
        - m^2 \delta \phi^* \phi
        - m^2 \phi^* \delta \phi
        \right].

\end{align*}
$$
Apply stokes theorem for manifolds to reorder $\delta \phi$

$$
\begin{gathered}
      \int \p_\mu \phi^* \p^\mu \delta \phi   d^4 x
    = \int \p^\mu (\p_\mu \phi^* \delta \phi) d^4 x
    - \int (\p^\mu \p_\mu \phi^*) \delta \phi d^4 x,
\\  = \oint \p^\mu \phi^* \delta \phi d^3 x
    - \int \p^\mu \p_\mu \phi^* \delta \phi   d^4 x
    = 0 - \int \p^\mu \p_\mu \phi^* \delta \phi   d^4 x,
\\ \vdots
\\      \delta S 
    = {1\over 2}\int d^4 x\ \left[    
        - \delta \phi^* \sq \phi
        - \sq \phi^* \delta \phi
             
        - m^2 \delta \phi^* \phi
        - m^2 \phi^* \delta \phi
        \right].
\end{gathered}
$$

We're free to set $\delta \phi = 0$, yielding

$$
\begin{gathered}
\delta S = -{1 \over 2} \int d^4 x\ \delta \phi^* \left[\sq \phi + m^2 \phi \right] = 0,
\\  \sq \phi + m^2 \phi = 0,
\end{gathered}
$$
with the other Euler-Lagrange equation being the same, conjugated.
