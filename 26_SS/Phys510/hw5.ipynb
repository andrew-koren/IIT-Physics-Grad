We are studying free complex scalar fields $\phi = \phi_1 + i\phi_2$ which exhibit $U(1)$ global phase symmetry, i.e. we can multiply by an arbitrary phase factor $e^{i\phi}$. I believe this means we can set our gauge as any $e^{i\phi(x)}$ as we did in class. We'll start from classical mechanics and work our way to quantization. The Lagrangian for the system is

$$
\def\L{\mathcal L}
\def\d{\partial}
\L = {1 \over 2} \d_\mu \phi^* \d^\mu \phi - {1 \over 2} m^2 \phi^* \phi.
$$
We would like to write this as $\L(\phi, \d^\mu\phi, x^\mu)$ and derive the Euler-Lagrange equations via least action principle. The action is $S = \int \L d^4 x$. This should collapse to Klein-Gordon-Fock for a special case

$$
\def\sq{\d^\mu\d_\mu}
\sq A^\mu + m^2 A^\mu = 0.
$$

Note we're using the minkowski metric from class with diagonal $(1,-1,-1,-1)$.
