```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import spherical_jn, spherical_yn

def calculate_sigma(ka_array, va, n_terms):
    """
    shamelessly generated! I did the math for hw#4
    Calculates sigma/a^2.
    ka is the independent variable.
    kpa (kappa * a) = ka + va
    """
    sigma_over_a2 = np.zeros_like(ka_array)
    
    for i, ka in enumerate(ka_array):
        # The relationship you specified: kappa*a = ka + v*a
        kpa = ka + va 
        total_sum = 0
        
        for l in range(n_terms):
            # Spherical Bessel j_l and j_{l-1}
            jl_ka = spherical_jn(l, ka)
            jlm1_ka = spherical_jn(l-1, ka) if l > 0 else np.cos(ka)/ka
            
            # Internal j_l and j_{l-1} at kappa*a
            jl_kpa = spherical_jn(l, kpa)
            jlm1_kpa = spherical_jn(l-1, kpa) if l > 0 else np.cos(kpa)/kpa
            
            # Neumann n_l and n_{l-1} at ka (External scattering happens at boundary a)
            # Note: The scattering occurs in the 'k' region, so n_l is evaluated at ka
            nl_ka = spherical_yn(l, ka)
            nlm1_ka = spherical_yn(l-1, ka) if l > 0 else -np.sin(ka)/ka
            
            # Boundary matching formula for tan(delta_l)
            # Numerator: k*j_l'(ka)*j_l(kpa) - kp*j_l'(kpa)*j_l(ka)
            # Denominator: k*n_l'(ka)*j_l(kpa) - kp*j_l'(kpa)*n_l(ka)
            # Using recursion: z*f_l' = z*f_{l-1} - (l+1)*f_l
            
            num = (ka * jlm1_ka * jl_kpa) - (kpa * jlm1_kpa * jl_ka)
            den = (ka * nlm1_ka * jl_kpa) - (kpa * jlm1_kpa * nl_ka)
            
            tan_delta = num / den
            sin2_delta = (tan_delta**2) / (1 + tan_delta**2)
            
            total_sum += (2*l + 1) * sin2_delta
            
        sigma_over_a2[i] = 4 * np.pi * total_sum
        
    return sigma_over_a2

# Parameters
ka_values = np.linspace(0.01, 2, 1000)
n_l = 30 # Number of partial waves (keep higher than max ka)
va_values = [0.0, 1.0, 3.0, 10.0, 100.0, 1000.0] # Different dimensionless 'va' offsets

plt.figure(figsize=(10, 6))

for va in va_values:
    sigma = calculate_sigma(ka_values, va, n_l)
    plt.plot(ka_values, sigma, label=fr'$va = {va}$ ($\kappa a = ka + {va}$)')

# Reference line: 4*pi*a^2 is a common low-energy scale, 
# but 2*pi*a^2 is the high energy limit.
plt.axhline(2 * np.pi, color='black', linestyle='--', alpha=0.3, label=r'Shadow Limit ($2\pi a^2$)')

plt.title(fr'Total Cross Section $\sigma/a^2$ with Fixed Offset $va$')
plt.xlabel(r'$ka$ (Dimensionless Momentum)')
plt.ylabel(r'$\sigma / a^2$')
plt.grid(True, which="both", ls="-", alpha=0.2)
plt.legend()
plt.show()
```
