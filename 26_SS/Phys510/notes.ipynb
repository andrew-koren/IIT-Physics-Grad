Quantum 1D chain

$$
\begin{align*}
    H       
    &= \sum_{M=-\infty}^{\infty} {p_M^2 \over 2 m} + K\left(x_M - x_{M+1} \right)
\\  \hat H  
    &= \sum_{M=-\infty}^{\infty} {\hat p_M^2 \over 2 m} + K\left(\hat x_M - \hat x_{M+1} \right)
\\  &= \sum_{M} \left[ 
      {\hbar \omega \over 4} 
      \left( \hat a_M - \hat a_M^\dagger\right)^2 
    + {K\hbar \over 2 m \omega} 
      \left(\hat a_M + \hat a_M^\dagger - \hat a_{M+1} - \hat a_{M+1}^\dagger \right)^2 
    \right]
\end{align*}
$$
Apply fourier transform 
$$
      \hat a_M = \sum_q \hat a_q e^{iqM}
\quad \hat a_M^\dagger = \sum_q \hat a_q e^{-iqM}
\\  \hat H = \sum_{M,q,q'} 
  {\hbar \omega \over 4} \left(\hat a_q - \hat a_{q'}^\dagger \right) \left( \hat a_{q'}-\hat a_{q}^\dagger \right) e^{i(q-q')M}
+ {K \hbar \over 2m\omega} 
        \left(1-e^{iq}\right)
        \left(1-e^{-iq'}\right)
        \left(\hat a_q + \hat a_{q'}^\dagger\right)
        \left(\hat a_{q'} + \hat a_{q}^\dagger \right)
        e^{i(q-q')M}
$$
There is some kind of identity we can apply in the infinite limit, allowing us to get rid of two sums simultaneously

$$
      \sum_M e^{i(q-q')M} 
    = \delta(q-q')
\\  \left(1 - e^{iq}\right) \left(1-e^{-iq}\right) = 2-2\cos q
\\    \hat H 
    = \sum_q 
    \underbrace{
        {\hbar \omega \over 4} 
        \left( \hat a_q - \hat a_q^\dagger \right)^2}
        _{\text{kinetic}}
    + 
    \underbrace{
        {2K \hbar \over m \omega}
        (1-\cos q) \left(\hat a_q + \hat a_q^\dagger \right)^2}
        _{\text{coupling}}

$$

Result: 
$$
\hat H 
    = \sum_k \left[ 
      p_k \left( 
          a_k^\dagger a_k 
        + a_k a_k^\dagger \right)
    + q_k \left( 
          a_k a_{-k} 
        + a_{-k}^\dagger a_k^\dagger \right) \right]
\\  p_k = {\hbar \omega \over 4} + {\hbar K \over 2M \omega} \left(1 - \cos k \right)
\\[0.25em]
    q_k =-{\hbar \omega \over 4} + {\hbar K \over 2M \omega} \left(1 - \cos k \right)
\\  p_{-k} = p_k \quad q_{-k} = q_k
$$

Bogoliubov Transformation - UV transformation

$$
\def\d{\dagger}
\\  a_k = u_k b_k + v_k b_{-k}^\d
\\  a_{-k} = v_k b_k^\d + u_k b_{-k}
\\  a_k^\d = u_k^* b_k^\d + v_k^* b_{-k}
\\  a_{-k}^\d = v_k^* b_k + u_k^* b_{-k}^\d
$$
Let's find u/v. First we want the same properties as c/a operators

$$
\def\l{\left}\def\r{\right}
\\  \l[b_k,b_k^\d\r] = 1
\\  \l[b_k,b_{-k}^\d\r] = 0
$$

Office hours question, remind me where creation/annihilation operators come from.

$$
\def\c#1#2{\l[#1,#2\r]}
      \c{a_k}{a_k^\d} 
\\  = \l(u_k  b_k    + v_k   b_{-k}^\d\r)\l(u_k^*b_k^\d+v_k^*b_{-k}   \r)
    + \l(u_k^*b_k^\d + v_k^* b_{-k}   \r)\l(u_k  b_k   +v_k  b_{-k}^\d\r)
\\  = |u_k|^2 \c{b_k}{b_k^\d} 
    + |v_k|^2 \c{b_{-k}^\d}{b_{-k}}
    + u_kv_k^* \c{b_k}{b_{-k}} 
    + u_k^* v_k \c{b_{-k}^\d}{b_k^\d}
\\  = 1
$$
According to our commutators above, this reveals
$$
|u|^2-|v|^2 = 1
$$
yielding the diagonal hamiltonian
$$
      \hat H
    = \sum_k \begin{align*}
        \Big\{& p_k \l[
            \l(u_k b_k^\d+v_k b_{-k}   \r)\l(u_k b_k    + v_k b_{-k}^\d\r)
        +   \l(u_k b_k   +v_k b_{-k}^\d\r)\l(u_k b_k^\d + v_k b_{-k}   \r)
        \r] 
\\      +
        & q_k \l[
            \l(u_k b_k      + v_k b_{-k}^\d \r)\l(u_k b_{-k}+v_k b_{k}^\d\r)
        +   \l(u_k b_{-k}^\d+ v_k b_k       \r)\l(u_k b_k^\d+v_k b_-k    \r)
        \r]\Big\}
    \end{align*} 
\\  = \sum_k \begin{align*}
        \Big\{ & p_k\l[
            u_k^2\l(b_k^\d b_k+ b_k b_k^\d\r)
        +   v_k^2\l(b_{-k}^\d b_{-k}+b_{-k}b_{-k}^\d\r)
        +   u_kv_k\l(b_k^\d b_{-k}^\d + b_{-k}b_{k}+b_{k}b_{-k}+b_{-k}^\d b_{k}^\d\r)
        \r] 
\\      +
        & q_k \l[
            u_k^2 \l(b_k b_{-k}          +b_{-k}^\d b_k^\d\r)
        +   v_k^2 \l(b_{-k}^\d b_{k}^\d  +b_k b_{-k}      \r)
        +   u_kv_k\l(b_k b_k^\d + b_{-k}^\d b_{-k} + b_{-k}^\d b_{-k} + b_k b_k^\d \r)
        \r]\Big\} \end{align*}
$$
Now just do some recombination to reveal our diagonal vs off-diagonal pieces
$$
      \hat H 
    = \sum_k \Big\{ 
        \underbrace{\l[
        \l(u_k^2+v_k^2\r)p_k+2u_kv_kq_k
        \r] \l(b_k^+b_k+b_kb_k^\d\r)}_{\text{diagonal}}
    + \underbrace{\l[
        2u_kv_kp_k + \l(u_k^2 + v_k^2 \r)q_k 
        \r]\l(b_kb_{-k}+b_{-k}^\d b_{k}^\d\r)}_{=0}
    \Big\}
$$
To cancel use
$$
    u_k = \cosh \lambda k;\quad v_k = \sinh \lambda k
\\  p_k \sinh^2\lambda k + q_k \cosh^2 \lambda k = 0
\\  \tanh^2 2\lambda k = {-q_k \over p_k}
$$
Solve the first part, which is $p_k \cosh 2\lambda k + q_k \sinh 2\lambda k$

$$
    \cosh 2\lambda k = {1 \over \sqrt {1 - \tanh^2 2\lambda k}} = { p_k \over \sqrt{p_k^2 -q_k^2}}
\\  \sinh 2\lambda k = {\tanh 2 \lambda k \over \sqrt{1 - \tanh^2 2\lambda k}} = -{q_k \over \sqrt{p_k^2 -q_k^2}}
\\  p_k \cosh 2\lambda k + q_k \sinh 2\lambda k = {\sqrt{p_k^2-q_k^2}}
\\  = \l\{ \l[ {\hbar \omega \over 4} + {\hbar K \over 2M \omega}\l(1-\cos k\r)\r]^2 
          -\l[-{\hbar \omega \over 4} + {\hbar K \over 2M \omega}\l(1-\cos k\r)\r]^2  \r\}^{1/2}
\\  = \l\{{\hbar \omega \over 2} {\hbar K \over M \omega}\l(1-\cos k\r) \r\}^{1/2}
    = {\hbar \over 2} \sqrt{K \over M} \l|\sin{K/2}\r|
\\    \hat H 
    = \hbar \sqrt{K \over M} \l|\sin{k/2}\r| \l(b_k^\d b_k + b_k b_k^\d \r)
$$
Or in absolutely canonical form
$$
\boxed{ \begin{matrix}
    \hat H = \sum_k \hbar \Omega(k) \l(b_k^\d b_k + {1 \over 2} \r)
\\  \Omega(k) = 2 \hbar \sqrt{K \over M} \l|\sin {k/2} \r|
\end{matrix}}
$$
