Since we only care about the relative values of the two bessel functions, we can enforce finite values / boundary conditions and be done without worrying about normalizing $C_l$.

Now, we can compute the cross-section analytically as

$$
\begin{align*}
\sigma  &= \int d\Omega \left| f(\theta)\right|^2,

\\      &= {2 \pi \over k^2} \int_0^\pi \sum_{l=0}^\infty \sum_{l'=0}^\infty A_l A_{l'}^* P_l(\cos \theta)P_{l'}(\cos \theta) d\theta.
\end{align*}
$$

We'll assume everything is good for now and we can swap summation/integration. In rigorous math, $P_l$ should take care of everything, but any valid answer will yield a non-infinite cross section anyways.
$$
\begin{align*}
\sigma&={2 \pi \over k^2}  \sum_{l=0}^\infty \sum_{l'=0}^\infty A_l A_{l'}^* 
        \int_0^\pi P_l(\cos \theta) P_{l'}(\cos \theta) d \theta,
\\    &={2 \pi \over k^2}  \sum_{l=0}^\infty \sum_{l'=0}^\infty A_l A_{l'}^* 
        {2\delta_{ll'} \over 2l+1}
       ={2 \pi \over k^2}  \sum_{l=0}^\infty (2l+1) {\tan^2 \delta_l \over 1+ \tan^2 \delta_l}.
\end{align*}
$$
Where $\delta_{ll'}$ is the kroneker delta. This means $\delta_l$ must converge to zero faster than $1 \over l$. Fortunately, for large $l$ we see $j_l$ shrink very quickly while $n_l$ grows ver quickly. Since $n_l$ is only in the denominator this will cause quick convergence.
