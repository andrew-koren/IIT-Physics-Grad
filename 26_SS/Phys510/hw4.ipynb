2. For
$$
V(r) = \lambda \delta(r-R)
$$
determine $\delta_0$ and the cross section in the limit $k\rightarrow 0$ when $\lambda = \pm |\lambda|$

Using $k_1 = {2m\lambda \over \hbar^2}$ we have

$$
\begin{matrix}
    u(r) = A \sin kr & r<R,
\\  u(r) = B \sin (kr+\delta_0) & r>R.

\end{matrix}
$$
We have a discontinuity $u'(R^+)-u'(R^-) = k_1u(R)$, yielding
$$
Bk\cos(kR+\delta_0) - Ak\cos(kR) = k_1 A\sin(kr).
$$
Dividing by the $u(r)$ (sort of like our logarithmic derivative) yields

$$
\begin{matrix}
    {Bk\cos(kR+\delta_0) \over u} - {Ak\cos(kR)\over u} = {k_1 A\sin(kr) \over u}
\\  k \cot (kR+\delta_0) - k\cot(kR) = g
\\  \cot(kR+\delta_0) = \cot(kR) + {g \over k}
\end{matrix}
$$
Apply trig identities to get
$$
\tan \delta_0 = -\frac{g\sin^2(kR)}{k+g\sin(kR)\cos(kR)}
$$

Using $k\rightarrow 0$ we can write out all our trig

$$
\begin{gathered}
    \delta_0 = -{k_0(kR)^2 \over k+k_0kR}
\\  = -k{2m\lambda^2 R^2 \over \hbar^2 + 2m\lambda R}
\end{gathered}
$$
yielding $\sigma$ as
$$
\begin{gathered}
\sigma_0 = {4\pi \sin^2 \delta_0 \over k^2} \approx {4\pi \delta_0^2 \over k^2}
\\ = 4\pi \left(2m\lambda R^2 \over \hbar^2 + 2m \lambda R\right)^2
\end{gathered}
$$
For positive lambda, this is just a soft sphere (becomes a hard sphere for $\lambda\rightarrow \infty$). If lambda is negative, it can behabe similar, but there is a resonance if $\hbar^2 = 2m|\lambda|R$, which will make our formula fail and need to use higher-order approximations.
