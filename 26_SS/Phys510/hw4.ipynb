$$
\begin{gathered}
    f(\theta) = {1 \over k} \sum_{l=0}^\infty \left(2l+1\right)P_l(\cos \theta) e^{i\delta_l} \sin(\delta_l),
\\  R_{kl}(r) = A_l j_l(kr) + B_l n_l (kr),
\\  \tan \delta_l = {B_l \over A_l}.
\end{gathered}
$$

Let's use $k$ inside the well and $\kappa$ outside to separate our two functions,
$$
      k^2 
    = {2m(E+V_0) \over \hbar^2},
\quad
      \kappa^2 
    = {2mE \over \hbar^2}.
$$
Since we only care about the relative values of $A_l$ and $B_l$, we don't need to worry about normalization, so we can enforce finite values / boundary conditions and be done.
