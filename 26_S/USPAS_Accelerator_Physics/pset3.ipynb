## Problem 3. Magnetic field from two straight wires

a) The field is just the superposition of the two wires. We can just use the infinite wire equation from class.

$$
|B| = {\mu_0 I \over 2\pi r} 
$$

<!-- By right hand rule, the bottom wire ($y=-a$) produces -->

To determine the direction, we need to use right-hand rule. See this very colorful diagram:



Everything is very clear now.

$$
    \vec R = - a \hat y + z \hat z
\\  R = \sqrt{a^2 + z^2}
\\  \hat R = {- a \hat y + z \hat z \over \sqrt{a^2 + z^2}}
\\  \hat B_{-\hat y} = -\hat x \times \hat R = {a \hat z + z \hat y \over \sqrt{a^2 + z^2}}
$$

For the $+a\hat y$ wire,
$$
\hat B_{+\hat y} = -\hat x \times \hat R = {-a \hat z + z \hat y \over \sqrt{a^2 + z^2}}
$$
yielding

$$
\vec B = |\vec B_{+\hat y}| \hat B_{+\hat y} + |\vec B_{-\hat y}| \hat B_{-\hat y} = {\mu_0 I \over \pi}{z \over \left(a^2 + z^2\right)} \hat y
$$

b) We will find ${d B_z \over dz} =0$

$$
    {d \over dz} {z \over \left( a^2 + z^2 \right)} = {a^2 - z^2 \over (a^2 + z^2)^2}=0 \rightarrow a= z
$$

c) See magnetic field lines



```python
a = 1
window = 2
resolution = 100
B_str = 1

y = np.linspace(-a*window, a*window, resolution)
z = np.linspace(-a*window, a*window, resolution)

Y, Z = np.meshgrid(y, z)

def B(y,z):
    length_p = np.sqrt((y-a)**2+z**2)
    length_m = np.sqrt((y+a)**2+z**2)
    Bp = B_str/length_p
    Bm = B_str/length_m

    Bp_vec = Bp*(z,(a-y))/(length_p)
    Bm_vec = Bm*(z,(-a-y))/(length_m)

    return Bp_vec + Bm_vec
By, Bz = B(Y,Z)

plt.streamplot(Y, Z, By, Bz)
plt.scatter((-a,a),(0,0), color='black')
plt.xlabel('y/a')
plt.ylabel('z')
```
