## Problem 1. Bending Magnets

a) There are 100 section that contain 8 sector magnets of length $\ell = 3.25$ m, so the total magnet length is $800\times 3.25 = 2.6\text{ km}$, yielding a fraction of $.65$.

b) $2\pi$ radians per $2.6 \text{ km}$, times $3.25$ m yields $7.86 \text{ mrad}$.

c) This comes out to $636$ m. Each bend has $\rho=\ell/\theta = 413$ m. These are different since the system isn't a continuous block of magnet and has some spacing for other elements, but there is still a total bend of $2\pi$ radians.

d) $B = {\gamma m c\beta \over q \rho} = {\gamma m c \over q \rho}\sqrt{1-{1 \over \gamma^2}}$, where $\gamma = {E \over mc^2}$

```python
from scipy.constants import c, e, m_p, mu_0 # kg

mc2 = 938 # MeV
E_initial = 50e3
E_final   = 300e3

gamma_i = E_initial/mc2
gamma_f = E_final/mc2

rho = 3.25/.00786 # m

B_i = gamma_i*m_p*c/(e*rho)*(1-1/(gamma_i**2))**(1/2)
B_f = gamma_f*m_p*c/(e*rho)*(1-1/(gamma_f**2))**(1/2)
print(B_i,B_f)
print(B_f*0.1/mu_0 *1e-6)
```

so 0.4T and 2.4T at injection and maximum, respectively

e) $\int B dl = \mu_0 N I$

$$
NI = {B l \over \mu_0} = 192643 \text{ amp-turns}
$$
