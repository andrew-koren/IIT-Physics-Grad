## Problem 1. Relativistic Diode

The velocity of electrons is constant $v\approx c$ which means charge is evenly distributed between plates. The current is simply

$$
J = \rho c
$$
so poissons equation can be written as
$$
{d^2 \phi \over dz^2} = -{J \over \epsilon_0 c}
$$
Where the rhs has no $z$ dependence. Integrating yields
$$
\begin{gathered}
{d \phi \over dz} = -{J \over \epsilon_0 c} z + C_0 = -E(0)
% \\  \phi(z) = -{J \over 2\epsilon_0 c} z^2 + C_0z + C_1
\end{gathered}
$$
Since $E(0)=0$ we have $C_0=0$. We integrate again and enforce $\phi(0)=0$ and $\phi(d)=V$
$$
\begin{gathered}
    \phi(z) = -{J \over 2\epsilon_0 c} z^2 + C_1
\\  \phi(0) = 0 \rightarrow C_1 = 0
\\  \phi(d) = V \rightarrow V = {Jd^2 \over 2 \epsilon_0 c}
\\  J = {2\epsilon_0 c V \over d^2}
\end{gathered}
$$

which results in $J$ depending linearly on $V$. See sketch below

$$
\begin{gathered}

\\  {p \over u} = \gamma \sqrt{1 - {1 \over \gamma^2}} {m_0 \over u} c= 505 \text{MeV/c/u}
\end{gathered}
$$

Thus for $q=78e$ we have

$$
B\rho = {{p/u} \over q}\times u = {505*238 \over 78} \left({10^{6} \over 3\times 10^8} \right) = 5.14 \text{ T m}
$$
where the last term handles unit conversion to tesla-meters (mev/c -> J/c)

```python
from scipy.constants import c

rest_mass = 931.5 # mev c^2 per nucleon
energy = 128 #mev
gamma = energy/rest_mass+1
beta = (1-(1/gamma)**2)**(1/2)
momentum = beta*gamma*rest_mass 
print('momentum per nucleon', momentum, 'MeV/c/u')
rigidity =momentum*238*1e6/(78*c)
print(rigidity)
```

## Problem 3. Magnetic chicane

a) trigonometry time. For the $\Delta L$ segment:

$$
\begin{gathered}
    \cos(\theta) = {\Delta L \over s}
\\  s_{\Delta L} = \Delta L \sec(\theta)
% \\  s(\theta) = 4L_B + 2 \Delta L \sec(\theta) + L_c + L_f
\end{gathered}
$$
For the path inside the dipole, we have bending radius $\rho = {p \over q B}= {L_B \over \theta}$

$$
\begin{gathered}
    \sin \theta = {L_B \over \rho}
\\  s = \rho \theta \neq 0
\\  s(\theta) = {L_B \theta \csc(\theta)}
\end{gathered}
$$

The total length is then

$$
4  {L_B \theta \csc(\theta)} + 2 \Delta L \sec(\theta) + L_c + L_f
$$

b) subtract the two

$$
\begin{gathered}
    \Delta s = 4L_B \left(\theta \csc(\theta) - 1\right) + 2\Delta L\left(\sec(\theta)-1 \right)
\\  \theta \csc(\theta) = 1 + \theta^2/6 + \dots
\\  \sec(\theta) = 1+ \theta^2/2 + \dots
\\  \Delta s = 4L_B (\theta^2/6 + \dots) + 2\Delta L \left( \theta^2 / 2 + \dots \right)
\\  \Delta s \approx \theta^2\left(\Delta L + {2 \over 3} L_B \right)
\end{gathered}
$$

c) Plugging in

$$
\begin{gathered}
    \Delta s = \left(\Delta L + 2 \over 3 L_B \right) \left( \theta_0 \over 1 + \delta \right)^2
\\  {1 \over 1 + \delta} = 1 - 2 \delta + 3 \delta^2 - 4 \delta^3 + \dots
\\  \Delta s \approx \theta_0^2\left(\Delta L + {2 \over 3} L_B \right) \left(1 - 2 \delta + 3 \delta^2 - 4 \delta^3 \right)
\end{gathered}
$$
Expanding yields $\Delta s_0 = \Delta_s$ from the previous problem, while the higher order terms are as expected
$$
\begin{gathered}
    s_{c} = \Delta L + {2 \over 3} L_B
\\  R_{56} = -2\theta_0^2 s_c
\\  T_{566} = 3 \theta_0^2 s_c = -{3 \over 2} R_{56}
\\  U_{5666} = -4 \theta_0^2 s_c = 2 R_{56}
\end{gathered}
$$
