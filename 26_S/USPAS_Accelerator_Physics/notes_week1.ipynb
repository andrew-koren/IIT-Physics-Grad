## Lecture 2: Transverse motion


Uniform $\vec B$ circular motion
$$
\begin{gathered}
    \vec F = {d \vec p d t} = q \left(\vec E + \vec v \times \vec B \right)
\\  \vec E = 0 \rightarrow \vec F = q \vec v \times \vec B
\\  \text{circular motion } F = {mv^2 \over R}
\end{gathered}
$$

Together yields the bending radius $\rho = R = {mv \over qB}$

The rigidity of a magnet is the momentum charge ratio $(B\rho) = {mv \over q}$. This yields $\rho = {(B \rho) \over B}$

Cyclotron motion is easy from the above. Betatron uses fulx induction.

$$
\begin{gathered}
    \Phi_B = \int_S \vec B_1 \cdot d \vec A
\\  \mathcal{E} = - {d \Phi_B \over dt}
\end{gathered}
$$

More general form of Faraday's law

$$
\begin{gathered}
\oint \vec E \cdot d\vec s = - {d \over dt} \int_S \vec B \cdot d\vec A,
& \quad &
\vec \nabla \times \vec E = - {\partial \vec B \over \partial t}
\end{gathered}
$$

Uniform $\vec B$ increasing in strength. Using integral form of Faraday's law & right hand rule, the increasing magnetic field will accelerate the particles along the path via EMF $\mathcal{E}$.

```python
def step_by_step_mul(operations: list):
    '''
    Matrix multiplication (xf) = [On,On-1,...,O2,O1](xi)
    '''
    O_r = operations[-1]
    for i in range(-1, -len(operations), -1):
            display(simplify(MatMul(*operations[0:i],O_r)))
            O_r = operations[i-1]*O_r
    result = expand(O_r)
    display(result)
    return result

result = step_by_step_mul([Defocusing,Drift,Focusing,Drift])
```

```python

S, C, Sp, Cp = symbols('S C S\' C\' ')

Twiss_Propogation_Matrix = Matrix([
    [C*C, -2*S*C, S*S],
    [-C*Cp, Sp*C + S*C, -S*Sp],
    [Cp*Cp, -2*Sp*Cp, Sp*Sp]
    ])

display(Twiss_Propogation_Matrix)

def transfer_to_twiss(transfer_matrix):
    '''Twiss 1D Parameter Evolution Matrix'''

    [[a,b],[c,d]] = transfer_matrix.tolist()
    parameters = {
        C: a,
        S: b,
        Cp: c,
        Sp: d
        }
    result = Twiss_Propogation_Matrix.subs(parameters)
    return result.copy()
```

Example: single rf in circular collider radius $\rho$.

$$

Wavevector $k = {2\pi \over L}$. 
Then $\omega_{\text{rf}} = k\beta c = {2\pi \over \tau} = h {2 \pi \over \tau}$, where $h$ is the harmonic number (integer?)



Increasing energy.

$$
\begin{align*}
        {d \omega \over \omega} 
&   = - {d \tau \over \tau}
