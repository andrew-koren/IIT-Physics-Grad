## Problem 1.

a) 

Nonrelativistic: Use energy gain from moving across capacitor

$$
\begin{align*}
  &E = T + U
\\&E(x=0) = U = q V
\\&E(x=d) = T = {1 \over 2} m v^2
\\&v = \sqrt{2 q V \over m}
\end{align*}
$$


Relativistic: Use $E = \gamma mc^2$ and $\gamma = {1 \over \sqrt{1 - \beta^2}}$

$$
\begin{align*}
  & q V + mc^2 = \gamma m c^2
\\& {q V\over mc^2} + 1 = \gamma 
% \\ & {q V\over mc^2} + 1 = \sqrt{1 \over 1 - \beta^2}
\\& \sqrt{1 - \beta^2} \left({q V\over mc^2} + 1 \right) = 1
\\& \sqrt{1 - \beta^2} = {1 \over \left({q V\over mc^2} + 1 \right)}
\\& {v^2 \over c^2}= 1-\left( {1 \over \left({q V\over mc^2} + 1 \right)} \right)^2
\\& {v^2}= c^2-\left( {c \over \left({q V\over mc^2} + 1 \right)} \right)^2
\\& v= \sqrt{c^2-\left( {c \over \left({q V\over mc^2} + 1 \right)} \right)^2}

% \\& v^2 = \left( {c \over \left({q V\over mc^2} + 1 \right)} \right)^2 - c^2
% \\& v = \sqrt{\left( {mc^3 \over{q V} + mc^2} \right)^2 - c^2}

\end{align*}
$$
which looks accurate, since the inside of the square root features a term minus the same term over some divisor $>1$.


b) Let's solve numerically



```python
from scipy.constants import m_e, e, c

# m_e kg
# e coulombs
# c m/s
print('nonrelativistic: V=10V and V=10kV (m/s and natural)')
print(f'{(2*e*10/m_e)**(1/2)} {(2*e*10/m_e)**(1/2)/c}')
print(f'{(2*e*10000/m_e)**(1/2)} {(2*e*10000/m_e)**(1/2)/c}')

print('\n')
get_v = lambda V: (c**2 - (c/(e*V/(m_e*c**2)+1))**2)**(1/2)
print('relativistic: V=10V and V=10kV (m/s and natural)')
print(get_v(V=10), get_v(V=10)/c)
print(get_v(V=10000), get_v(V=10000)/c)
```

(c) convert MeV to joules

$$
    E = \gamma mc^2 = T + mc^2
\\  \gamma = {T + mc^2 \over mc^2}
\\  \beta = \sqrt{1 - {1 \over \gamma^2}}
$$

```python
from scipy.constants import m_p
# E = KE + mc^2
E = 100e6 * 1.602e-19 + m_p*c**2 # joules
gamma = E/(m_p*c**2)
beta = (1-1/(gamma**2))**(1/2)
print('gamma, beta, m/s')
print(gamma, beta, beta*c)
```

d) Protons and neutrons have the same mass for our purposes. Since kinetic energy scales linearly with mass, this allows for rescaling $\gamma$ as

$$
    \gamma = {T + n\times m_p c^2 \over n \times m_pc^2}
\\  = {n{T \over n} + n\times m_p c^2 \over n \times m_pc^2}
\\  = {{T \over n} + m_p c^2 \over m_pc^2}
$$
And $\beta$ is fixed the same way as $\gamma$. Then all heavy nuclei $\beta$ are scaled equivalently by kinetic energy over atomic mass number. Since this is equivalent for all heavy nuclei, specifying MeV/nucleon of an RF cavity structure always describes the energy of the objects it accelerates.

## Problem 2.

(a) We need to compute $\gamma$, let's convert from total energy per nucleon to total energy

$$
\begin{gathered}
    E = \gamma m c^2
\\  m = 197 \times 0.93113 \text{ GeV / c}^2 = A \times m_p
\\  E = A {E \over A}
\\  A = 197
\\  \gamma = {A \times {E \over A} \over m c^2}
\\  = {{E \over A} \over m_p}
\end{gathered}
$$

```python
E_inj = 10.5
E_col =  100
m_p = 0.93113 # avg gold nucleon energy

gamma_inj = E_inj / m_p
gamma_col = E_col / m_p

v_inj = (1 - (1/gamma_inj**2))**(1/2) * c
v_col = (1 - (1/gamma_col**2))**(1/2) * c

# print(v_inj/c, v_col/c)
circumference = 3834
f1 = v_inj/circumference
f2 = v_col/circumference
print('Injection, Collision, Difference (Hz)')
print(f1, f2, f2-f1)
```
