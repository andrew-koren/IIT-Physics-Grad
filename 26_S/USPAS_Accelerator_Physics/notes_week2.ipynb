## Lecture 10: Fundamentals of RF Cavities

Note Helmut uses $\mu_0, \epsilon_0 = 1$. Seeh Helmut 18

$$
\begin{gathered}
    \vec D = \epsilon \vec E,
\qquad
    \vec B = \mu \vec H
\end{gathered}
$$Source free region / vacuum

$$
\begin{darray}{cc}
    E = E_0 e^{i\omega t},
&   \vec J = 0,
\\  \vec D = \epsilon_0 \vec E,
&   \vec B = \mu_0 \vec H,
\\  \vec \nabla \cdot \vec E = 0,
&   \vec \nabla \cdot \vec B = 0,
\\  \vec \nabla \times \vec H = i\omega \epsilon_0 \vec E,
&   \vec \nabla \times \vec E = - i \omega \mu_0 \vec H,
\end{darray}

$$

Curl to get wave equations.

$$
\begin{gathered}
    \vn \times \l( \vn \times \vec A \r) = \vn\l(\vn\cdot\vec A\r) - \nabla^2\vec A
\\  \nabla^2 \vec H = -\epsilon_0\mu_0 \omega^2 {\partial^2 \vec H \over \partial t^2}
\\  \nabla^2 f = {1 \over V^2}{\partial^2 f \over \partial t^2}
\\  \nabla^2 \vec E = - \epsilon_0 \mu_0 \omega^2 {\partial^2 \vec E \over \partial t^2}
\end{gathered}
$$
velocity in vacuum $\vec E \perp \vec B \perp \vec k$. $V_{\text{ph}} = c = {1 \over \sqrt{\epsilon_0 \mu_0}}$.

$$

$$

Photoemission, now the laser penetrates into the material, or doesn't and gives the electron extra energy.

$$
I(x,h\nu) = I_0 (h \nu) \big[ 1 - \underbrace{R(h\nu)}_{\text{reflectivity}} \big]e^{-\alpha(h\nu)x},
$$
where $\alpha$ is the absorption coefficient and $I_0$ is the incoming beam amplitude.

Example: copper $\phi=4.6\text{eV}$ needs $\lambda = 270\text{nm}$ uv light.

Quantum yield/efficiency.

$$
QY(h\nu) = \frac{{\alpha_{\text{PE}}\over \alpha}P_E}{1+L/\alpha }
$$
where $\alpha_{\text{PE}}$ describes the radiation's ability to excite electrons and $P_E$ describes the probability an electron will surface with sufficient energy.

Results in slightly different emmitance once work function is modified.

$$
\epsilon_{\text{photo}} = \sigma_x \sqrt{h \nu - \phi_{\text{eff}} \over 3mc^2}
$$

Solenoid focusing. Cylindrical coordinates with azimuthal symmetry

See Hemut 6,20
$$
\begin{gathered}
        \vec B 
    =   B_r(r,z)\hat r + B_z(r,z) \hat z
\\      B_z(r,z)
    =   B(z) - {r^2 \over 4} B''(z) + \dots
    \approx B(z)
\\      B_r(r,z)
    = - {r \over 2} B'(z) + {r^3 \over 16} B^{(3)}(z) + \dots
    \approx -{r \over 2} B'(z)
\end{gathered}
$$

If we start with a paraxial beam $\vec v = v_z \hat z$

$$
\vec v \times \vec B \sim \hat z \times - \hat r = -\hat \theta
$$
So there is a kick $\Delta v_\theta = r_0 {q B' \over 2 \gamma m}$, with $r_0$ radial distance at entrance.

This will orbit around a radius $R = {r_0 \over 2}$. Cyclotron frequency $\omega_c = {qB_0 \over \gamma m}$



The $\Delta V_\theta$ kick is in the opposite direction, causes focusing.
