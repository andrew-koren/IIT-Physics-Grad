$$
\begin{gathered}    
\hat B = \pm' \hat z \times \hat r_{\text{sep}}
\\ \begin{gathered}
    |B_c| = {\mu_0 I_0 \over 2 \pi c}

&   |B_b| = {\mu_0 I_0 \over 2\sqrt{2} \pi b}
\\  \hat r_{\text{sep},b} = - \hat x
&   \hat r_{\text{sep},c} = {\hat x \mp \hat y \over \sqrt 2}
\\  \hat B_c= \mp' \hat y
&   \hat B_{b\pm}= {\pm' \hat y \pm'\pm \hat x \over \sqrt 2}
\end{gathered}
\end{gathered}
$$

We can control the $\pm'$ for each wire individually, but cancelling the $\hat x$ component of the two $\vec B_b$ wires requires alignment. This yields

$$
\begin{gathered}
    \vec B_{b-} + \vec B_{b+} = \pm' {\mu_0 I_0 \over 2 \pi b} \hat y
\\  \vec B_c = \mp' {\mu_0 I_0 \over 2\pi c} \hat y

$$
