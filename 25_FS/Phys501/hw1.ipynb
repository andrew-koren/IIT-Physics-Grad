3. Show that $\braket{f-\bar{f}, \phi_i}=0$ for all $\phi_i$

Remember linearity $\braket{f-\bar{f}, \phi_i} = \braket{f, \phi_i} - \braket{\bar{f}, \phi_i}$

In proposition 0.3.2, we define <!-- pg 25, pdf pg 41 -->

$$
c_i = \hat c_i := \frac{\braket{f,\phi_i}}{\|\phi_i\|^2}
$$

Since $(\phi_1, \dots,\phi_n)$ form an orthogonal set, $\|\phi_i \|^2 = 1$ and
$$
\hat c_i = \braket{f, \phi_i}
$$

We can expand $\bar{f}$ and find the inner product of each orthogonal component

$$
\begin{align*}
     \braket{\bar f, \phi_i} 
  &= \braket{\hat c_1 \phi_1 + \dots + \hat c_i \phi_i + \dots \hat c_n \phi_n, \phi_i}
\\&=  \braket{\hat c_1 \phi_1, \phi_i} + \dots
    + \braket{\hat c_i \phi_i, \phi_i} + \dots
    + \braket{\hat c_n \phi_n, \phi_i}
\\&=  \hat c_1\braket{\phi_1, \phi_i} + \dots
    + \hat c_i\braket{\phi_i, \phi_i} + \dots
    + \hat c_n\braket{\phi_n, \phi_i}
\\&= 0 + \dots + \hat c_i + \dots + 0
\end{align*}
$$

Plugging in this result, we show orthogonality for any $\phi_i$
$$
   \braket{\bar{f}, \phi_i} = \hat c_i = \braket{f, \phi_i} 
\\ \braket{f - \bar{f}, \phi_i}
  =\braket{f, \phi_i} - \braket{\bar{f}, \phi_i} 
  =\hat c_i - \hat c_i = 0
\\ \braket{f - \bar{f}, \phi_i} = 0 \quad \square
$$

16 Fourier sine series 
<br> a) $f(x) = x, 0 \leq x \leq L$
<br> b) $f(x) = x^2, 0 \leq x \leq L$


The Fourier sine series is
$$
\sum_{n=1}^\infty B_n \sin \frac{n\pi x}{L} \\
B_n = \frac{2}{L} \int_0^L f(x) \sin\frac{n\pi x}{L} dx
$$

I'm going to use sympy to evaluate these integrals.

a) 
$$
    B_n = \frac{2}{L} \int_0^L x\sin\frac{n\pi x}{L} dx  
\\  = \frac{2L(\sin(n\pi)-n\pi \cos(n\pi))}{n^2\pi^2}
\\  = \frac{-2L(-1)^n}{n\pi}
$$

b)
$$
    B_n = \frac{2}{L} \int_0^L x^2\sin\frac{n\pi x}{L} dx
\\  = \frac{2 L^{2} \left(- \pi^{2} n^{2} \cos{\left(\pi n \right)} + 2 \pi n \sin{\left(\pi n \right)} + 2 \cos{\left(\pi n \right)} - 2\right)}{\pi^{3} n^{3}}

\\  = \frac{2 L^{2} \left(- \pi^{2} n^{2} (-1)^n + 2 (-1)^n - 2\right)}{\pi^{3} n^{3}}

\\  = \frac{2 L^{2} \left(\left(- \pi^{2} n^{2} + 2\right) (-1)^n - 2\right)}{\pi^{3} n^{3}}
$$


```python
display((2/L * integrate(x*sin(n*pi*x/L), (x, 0, L))).simplify())
display((2/L * integrate(x**2*sin(n*pi*x/L), (x, 0, L))).simplify())
```
