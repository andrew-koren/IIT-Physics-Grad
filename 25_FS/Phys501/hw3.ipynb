4. Parabolic coordnates
$$
    x = \xi \eta
\\  y = \frac{1}{2} (\xi^2 - \eta^2)
$$

Find Lame coefficients and Laplacian $\nabla^2 f$. Double-check orthogonality

$$
\def\d#1#2{\frac{\partial #1}{\partial #2}}
    g_{\xi \eta} = \d{x}{\xi}\d{x}{\eta} + \d{y}{\xi}\d{y}{\eta}
$$
Considering the next problem, let's set up sympy

```python
from sympy import *

def get_metric(xyz_, abc_):
    lame = S.Zero
    for x_i in xyz_:
        term = S.One
        for a_i in abc_:
            term *= diff(x_i, a_i)
        lame += term

    return lame

def get_h(xyz_, abc_):
    h_i = [ 
        sqrt(sum(diff(x_i, a_i)**2 for x_i in xyz_)) for a_i in abc_ 
        ]

    return h_i

g, h_i, xi, eta = symbols('g_xi_eta h_i xi eta', real=True)


x = xi*eta
y = 1/2 * (xi*xi - eta*eta)

display(Eq(g, get_metric((x,y), (xi, eta))))

for i, h_i in enumerate(get_h((x,y), (xi, eta))):
    h_symbol = symbols(f'h_{i+1}')
    display(Eq(h_symbol, h_i))
```

$g_{\eta \xi} = 0 $ garuntees orthogonality. Since $h_1 = h_2$, this simplifies down to
$$
\renewcommand{\d}[2]{\frac{\partial #1}{\partial #2}}

\nabla^2 = \frac{1}{\eta^2 + \xi^2} \left[\d{}{\eta^2} + \d{}{\xi^2} \right]
$$

5. Bispherical system

$$
    x = \frac{a \sin  \xi \cos \phi}{\cosh \eta - \cos \xi}
\\  y = \frac{a \sin  \xi \sin \phi}{\cosh \eta - \cos \xi}
\\  z = \frac{a \sinh \eta         }{\cosh \eta - \cos \xi}
$$

```python
a, phi = symbols('a phi', real=True)

denom = cosh(eta) - cos(xi)
x = a * sin (xi) *  cos(phi)  / denom
y = a * sin (xi) *  sin(phi)  / denom
z = a * sinh(eta)             / denom

# a isn't a coordinate
assert not simplify(get_metric((x,y,z), (xi, phi)))
assert not simplify(get_metric((x,y,z), (xi, eta)))
assert not simplify(get_metric((x,y,z), (phi, eta)))

coords = (xi, eta, phi)
h = get_h((x,y,z), coords)
h_clean = []
for i, h_i in enumerate(h):
    h_symbol = symbols(f'h_{coords[i]}')
    h_simp = simplify(expand(h_i))
    h_simp = trigsimp(h_simp, recursive = True)
    display(Eq(h_symbol, h_simp))
    h_clean.append(h_simp)
```

Trigsimp does the trick, now our answer for $h_i$ match wikipedia.
