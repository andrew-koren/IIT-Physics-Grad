a) Use polynomial series $y = \sum_{n=0}^\infty a_n x^n $

$$
   y'  = \sum_{n=0}^\infty n a_n x^{n-1}
\\ y'' = \sum_{n=0}^\infty n (n-1) a_n x^{n-2}
       = \sum_{n=2}^\infty n (n-1) a_n x^{n-2}
\\ \underbrace{\sum_{n=2}^\infty n (n-1) a_n x^{n-2}}_{n \rightarrow n+2} - \sum_{n=0}^\infty a_n x^n = 0
\\ \sum_{n=0}^\infty (n+2) (n+1) a_{n+2} x^{n} - \sum_{n=0}^\infty a_n x^n = 0
\\ \sum_{n=0}^\infty  \left[(n+2) (n+1) a_{n+2} - a_n \right] x^n= 0
\\ (n+2) (n+1) a_{n+2} - a_n = 0
\\ a_{n+2} = \frac{a_n}{(n+2)(n+1)}
\\ a_{n} = \frac{a_{n-2}}{(n)(n-1)}
$$
The $a_n$ terms are found to be $a_0$ and $a_1$ divided by factorial
$$
\begin{matrix}
& a_{2} = \frac{a_0}{1\cdot 2}
& a_{3} = \frac{a_1}{2\cdot 3}
& a_{4} = \frac{a_0}{(1\cdot 2)(3\cdot 4)}
& a_{5} = \frac{a_1}{(2\cdot 3)(4\cdot 5)}
\\&& a_{2n} = \frac{a_0}{(2n)!}
& a_{2n+1} = \frac{a_1}{(2n+1)!}
\end{matrix}
$$
Giving an infinite series solution for $y$ as
$$
y(x) = a_0\sum_{n=0}^\infty \frac{x^{2n}}{(2n)!} + a_1 \sum_{n=0}^\infty \frac{x^{2n+1}}{(2n+1)!}
$$
where $a_0$ and $a_1$ depend on inital values. This result is equal to the hyperbolic sine/cosine expansions
