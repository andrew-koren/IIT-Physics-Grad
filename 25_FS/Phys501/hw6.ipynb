### 3.5.4


We might be able to make this one out of the three solution types, but those are all optimized around $u=0$ not $u_z=0$. It'll be smarter to derive our own Neumann or Mixed case instead. From the laplace operator in cylinderical coordinates and separation, we have

$$
      {R'' + (1/r)R' \over R} 
    + {1 \over r^2} \underbrace{{\Phi'' \over \Phi}}_{-m^2}
    +               \underbrace{Z''     \over Z    }_{-\lambda^2}
    = 0
\\  Z'' = - \lambda z
\\  Z(z) = A \sin(\lambda z) + B \cos(\lambda z)
\\  Z(0) = 0 \rightarrow B = 0
\\  Z'(L) = 0 
    \rightarrow \lambda L = {(2n-1)\pi \over 2}
\\  \rightarrow \lambda = {(2n-1)\pi \over 2L}
\\  Z(z) = \sin \left((2n-1)\pi z \over 2L \right)
\\  \Phi(\phi) = A \sin(m\phi) + B \cos(m\phi),
$$
which results in the modified bessel function
$$
R'' + {R' \over r} - \left( \left((2n-1)\pi \over 2L \right)^2 + {m^2\over r^2}\right) R = 0
\\
R(r) = I_m \left((2n-1)\pi r \over 2L \right)
$$

Giving us the perfect custom solution (after removing angular dependence)

$$
u(r,\phi,z) = \sum_{n=1} 
    A_n I_0 
    \left((2n-1)\pi r \over 2L \right)
    \sin \left((2n-1)\pi z \over 2L \right)
$$
and, all we need is a sine expansion to get our boundary condition
