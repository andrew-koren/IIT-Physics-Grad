13. Derive equation of motion and $v(m)$

The change in momentum of the rocket is
$$
\frac{dp}{dt} = \frac{dp_{gas}}{dt} + f
$$
The change in momentum due to exhaust must factor in the changing mass and velocity, and our result uses exhaust relative velocity.
$$
\frac{d}{dt} m_{gas}v_{gas} = (v-v') \frac{dm_{gas}}{dt} + m_{gas} \frac{d(v-v')}{dt}
\\ = (v-v') \frac{dm_{gas}}{dt} + m_{gas} \frac{dv}{dt} - m_{gas}\frac{dv'}{dt}
$$
A reasonable assumption is that the ejected gas relative velocity is constant, especialy with constant loss of mass rate.

$$
\frac{dm_{gas}}{dt} = \frac{dm}{dt}
\\ \frac{d}{dt} m_{gas}v_{gas} = (v-v')  \frac{dm}{dt} + m_{gas} \frac{dv}{dt}
$$
We're also only concerned with the effects of the gas on the rocket, which do not depend on what happens to the mass after it leaves the ship, sending $m_{gas} \frac{dv}{dt}$ to zero. <!--One way to understand this well is our zero-friction assumption, where air resistance would really eat up this component of momentum. -->
$$
\\ \frac{d}{dt} m_{gas}v_{gas} = (v-v')  \frac{dm}{dt} 
$$


<!-- Momentum is not conserved due to the gravitational field, so we have
$$
F = -mg = \frac{d}{dt} mv
$$
giving us
$$
\\ -mg = v \frac{dm}{dt} + m \frac{dv}{dt}
\\ m \frac{dv}{dt} = - v \frac{dm}{dt} - mg
$$ -->

Since the rocket has variable velocity and mass, we must take the derivative of both for $dp/dt$

$$
\frac{dp}{dt} = m\frac{dv}{dt} + v\frac{dm}{dt} 
\\= v\frac{dm}{dt} - v'\frac{dm}{dt} +f
\\ m\frac{dv}{dt} = -v'\frac{dm}{dt}  - mg
\\ \frac{dv}{dt} = -\frac{v'}{m} \frac{dm}{dt} - g
\\[1em] \frac{dv}{dm} \frac{dm}{dt}  = - \frac{v'}{m} \frac{dm}{dt} - g
\\      \frac{dv}{dm}    = - \frac{v'}{m}  - g\frac{1}{\dot m}
\\ \int_{m_0}^m \frac{dv}{dm} dm = - v'\int_{m_0}^m \frac{1}{m} -\frac{g}{\dot m} \int_{m_0}^m  dm
\\ v(m) = v_0 -v' \ln \frac{m}{m_0} - \frac{g}{\dot m} (m - m_0)
$$

Let's plug in. While I could do some inversion and get $m$ as a function of escape velocity, everything is approximated here anyways so I might as well go the other direction.
$$
\\ v_0 = 0
\\ v' = 2100 \text{ m/s}
\\ \dot m = -1/60 m_0
\\ m = m_0/301
\\ v(m) = 2100 \ln 301  - (9.8)(60)(-\frac{300}{301} )
\\ = 12571 \text{m/s}
$$

Escape velocity is a bit above 11 km/s so this checks out.

22. Glad I did this in my undergrad. There's no mention of gravity, but it'd be a pretty boring system without it, and minor pedantics would say it isn't a pedulum without restorative force.


$$
   L = \frac{m_1}{2}v_1^2 +\frac{m_2}{2} v_2^2 + \frac{I}{2}(\dot \theta_1^2 + \dot \theta_2^2)
\\   = \frac{m}{2}(\dot x_1^2 + \dot y_1^2)+ \frac{m}{2}(\dot x_2^2 + \dot y_2 ^2) + \frac{I}{2}(\dot \theta_1^2 + \dot \theta_2^2)
$$

Fortunately our point masses have simpler kinetic energy. Let's parameterize out everything with $\theta_1$ and $\theta_2$. The postions of the two masses are given by

$$
   \begin{pmatrix} x_{1} \\ y_1 \end{pmatrix} = l_1 \begin{pmatrix} \sin \theta_1 \\ -\cos \theta_1 \end{pmatrix}
\\ \begin{pmatrix} x_{2} \\ y_2 \end{pmatrix} = \begin{pmatrix} x_{1} \\ y_1 \end{pmatrix} + l_2\begin{pmatrix} \sin \theta_2 \\ -\cos \theta_2 \end{pmatrix}
\\ = \begin{pmatrix} l_1\sin \theta_1 + l_2\sin \theta_2 \\ -l_1\cos \theta_1 - l_2\cos \theta_2 \end{pmatrix}
$$

Next up is differentiation. Rather than writing this all out, maybe we should make use of our friend sympy

```python
from sympy import *
from sympy.physics.mechanics import dynamicsymbols

# coordinate system with dθ/dt
th1, th2 = dynamicsymbols('theta1 theta2') # will be time-dependent
th1d, th2d = diff(th1), diff(th2)

l1, l2, m1, m2, g = symbols('l1 l2 m1 m2 g')

# p1 & p2
x1 = l1 * sin(th1)
y1 = -l1 * cos(th1)
x2 = x1 + l2 * sin(th2)
y2 = y1 - l2 * cos(th2)

# velocities
vx1, vy1, vx2, vy2 = [diff(x_i, 't') for x_i in [x1, y1, x2, y2]]
```

```python
T = (1/2) * m1 * (vx1**2 + vy1**2) + (1/2) * m2 * (vx2**2 + vy2**2)
V = m1*g*y1 + m2*g*y2

L = T - V

# print(latex(simplify(expand_trig(L))))
a = Eq(diff(L, th1) - diff(diff(L, th1d), 't'), 0).simplify()
b = Eq(diff(L, th2) - diff(diff(L, th2d), 't'), 0).simplify()

#display((simplify(solve(a, diff(th1d, 't'))[0])))
#print(latex(simplify(solve(b, diff(th2d, 't'))[0])))
#a.replace(diff(th2d, 't'), solve(b, diff(th2d, 't'))[0])

print('T:')
display(simplify(T))
print('V:')
display(simplify(V))
```
