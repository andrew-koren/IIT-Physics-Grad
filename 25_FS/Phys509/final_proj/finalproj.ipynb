$$
    \hat H = \hat H_{\text{field}}+\hat H_{\text{atom}}+\hat H_{\text{interaction}}
\\[0.5em]
    \begin{align*}
    & \hat H_{\text{field}}       = \hbar \omega_c \hat a^\dagger \hat a
\\  & \hat H_{\text{atom}}        = \hbar \omega_a \frac{\hat \sigma_z}{2}
\\  & \hat H_{\text{interaction}} = \frac{\hbar \Omega}{2} \hat E \hat S
\end{align*}
$$

Where $\hat E = E_{\text{ZPF}} \left( \hat a + \hat a^\dagger \right)$ and $\hat S$ is the polarization $\hat S = \hat \sigma_+ + \hat \sigma_-$. The "atom" can be any two level system, with ground state $\ket g $ and excited state $\ket e$. The $\hat \sigma$ matricies can be written as projections

$$
\begin{align*}
\\  & \hat \sigma_z = \ket{e}\bra{g}
\\  & \hat \sigma_+ = \ket{g}\bra{e}
\\  & \hat \sigma_- = \ket{e}\bra{e} - \ket{g}\ket{g}
\end{align*}
$$

Heisenberg vs Schrodinger picture?

To develope a time-dependent perturbation interaction, first consider the simple unperturbed hamiltonian

$$
\bar H = \hat H_{\text{field}} + \hat H_{\text{atom}}
$$

In this case, the hamiltonian is seperable

$$
\\   \bar H \ket E = E \ket E
\\[0.5em]
    \hat H_{\text{field}} \ket{n} = \hbar \omega_c \hat a^\dagger \hat a \ket{n}
    =\hbar \omega_c n \ket{n}
\\[0.5em]
    \hat H_{\text{atom}} \ket{\pm }  = \frac{\hbar \omega_a}{2} \hat \sigma_z \ket{g} = \pm \frac{\hbar \omega_a}{2} \ket{\pm}
\\[0.5em]
    \bar H \ket{n} \otimes \ket{\pm} = \hbar \left( \omega_c n \pm \frac{\omega_a}{2} \right) \ket{n} \otimes \ket{\pm}
$$
Where $\ket{-} = \ket{g}$ and $\ket{+} = \ket{e}$

Next, the time-independent perturbation (Schrödinger representation)

$$
    V = \frac{\hbar \Omega E_{\text{ZPF}}}{2} \left( \hat a + \hat a^\dag\right)\left(\hat \sigma_+ + \hat \sigma_-\right)
$$
