## The Jaynes-Cummings Hamiltonian 
is actually a simplifciation of another Hamiltonian, the Rabi Hamiltonian

$$
    \hat H_{\text{Rabi}} = \hat H_{\text{field}}+\hat H_{\text{atom}}+\hat H_{\text{interaction}}
\\[0.5em]
    \begin{align*}
    & \hat H_{\text{field}}       = \hbar \omega_c \hat a^\dagger a
\\  & \hat H_{\text{atom}}        = \hbar \omega_a \frac{\hat \sigma_z}{2}
\\  & \hat H_{\text{interaction}} = \hbar \Omega \left(\hat a^\dagger + \hat a\right) \left(\hat \sigma_- + \hat \sigma_+ \right)
\end{align*}
$$

Where the intearction represents the coupling of the electric field operator $(\hat a + \hat a^\dagger)$ and polarization operator $(\hat \sigma_+ + \hat \sigma_-)$, essentially a dipole interaction. I've chosen to write the coupling strength as $\Omega$, although literature commonly uses $g$ or other symbols to bundle up constants. 

To get to the Jaynes-Cummings Hamiltonian, we assume the Rotating Wave Approximation (RWA), which suggests that interaction terms with energy not conserved go to zero. Expanding out the interaction term, we get

$$
    \left(\hat a^\dagger + \hat a\right) \left(\hat \sigma_- + \hat \sigma_+ \right) 
    = \hat a^\dagger \hat \sigma_- + \hat a^\dagger \hat \sigma_+ + \hat a \hat \sigma_- + \hat a \hat \sigma_+
$$

The terms $\hat a^\dagger \hat \sigma_+$ and $\hat a \hat \sigma_-$ represent non-energy-conserving processes where excitations are simultaneously created or simultaneously destroyed in both the atom and the cavity. The RWA discards these terms, leaving only the energy-conserving terms. This yields the Jaynes-Cummings interaction 

$$
\hat H_{\text{int}} = \hbar \Omega (\hat a^\dagger \hat \sigma_- + \hat a \hat \sigma_+)
$$
which describes the coherent exchange of a single excitation between the field and the atom. The "atom" can actually be any two level system, with ground state $\ket g $ and excited state $\ket e$. For the two-level sytem, the $\hat \sigma$ matricies can be written as projections

$$
\def\bra#1{\langle#1|}
\def\ket#1{|#1\rangle}
\begin{align*}
\\  & \hat \sigma_z = \ket{e}\bra{e} - \ket{g}\bra{g}
\\  & \hat \sigma_+ = \ket{e}\bra{g}
\\  & \hat \sigma_- = \ket{g}\bra{e}
\end{align*}
$$

 This Hamiltonian thus gives $\hbar \omega_a$ is the harmonic oscillator energy gap, while $\hbar\omega_c$ is the two-state energy gap.

## Time-Independent Perturbation theory

Before I realized that the Jaynes-Cummings Hamiltonian is block-diagonal in the Total Excitation basis, I also computed the time-independent perturbation. At first, I thought I had to throw this result out, but then I realized it is a great example for comparing perturbation theory with a solvable system.

To develope a time-independent perturbation interaction, first consider the simple unperturbed hamiltonian:

$$
\bar H = \hat H_{\text{field}} + \hat H_{\text{atom}}
$$

In this case, the eigenstates are a product basis

$$
\\   \bar H \ket E = E \ket E
\\[0.5em]
    \hat H_{\text{field}} \ket{n} = \hbar \omega_c \hat a^\dagger a \ket{n}
    =\hbar \omega_c n \ket{n}
\\[0.5em]
    \hat H_{\text{atom}} \ket{\pm }  = \frac{\hbar \omega_a}{2} \hat \sigma_z \ket{\pm} = \pm \frac{\hbar \omega_a}{2} \ket{\pm}
\\[0.5em]
    \bar H \ket{n} \otimes \ket{\pm} = \hbar \left( \omega_c n \pm \frac{\omega_a}{2} \right) \ket{n} \otimes \ket{\pm}
$$
Where $\ket{-} = \ket{g}$ and $\ket{+} = \ket{e}$. 

In time-independent perturbation (Schrödinger representation), we have the perturbation

$$
\begin{align*}
       V 
    &= \hbar \Omega (\hat a^\dagger \hat \sigma_- + \hat a \hat \sigma_+)
\\  &= \left(\hbar \Omega \hat a^\dagger \ket{g}\bra{e}\right) 
     + \left(\hbar \Omega \hat a \ket{e}\bra{g}\right).

\end{align*}
$$
It should be aparent that the first-order correction is zero, but the second-order correction is

$$
      E_{s}^{(2)} 
    = \sum_{k \neq s} \frac{|\bra{k}V \ket{s}|^2}{(E_s^{(0)}-E_k^{(0)})}
$$

where $E_i^{(0)}$ is the energy of the unperturbed state, and each state $k$ or $s$ has a combination of harmonic oscillator state $\ket n$ and atom state $\ket \pm$. We'll check the correction for the state $s_+=\ket{n,+} = \ket{n} \otimes \ket{e}$ and $s_- = \ket{n} \otimes \ket{g}$

$$
\def\braket#1{\langle #1 \rangle}
\\  \bra{k_1} V \ket{s_+} = \braket{k_1|n+1,g} \hbar \Omega \sqrt{n+1} 
\\  \bra{k_2} V \ket{s_-} = \braket{k_2|n-1,e} \hbar \Omega \sqrt{n} 
$$
which are of course nonzero only for $\bra{k_1} = \bra{n+1,g},\ \bra{k_2}=\bra{n-1,e}$. Thus, the corrected energies are

$$
\\  E_{n,e}^{(2)} 
    = \frac {|\sqrt{n+1} \hbar \Omega|^2 }
            {\hbar\left(
                  \omega_c n + {\omega_a \over 2} 
                - (\omega_c(n+1) - {\omega_a \over 2} )
            \right)}
    = \frac {\hbar \Omega^2 (n+1) }
            {\omega_a - \omega_c} 
\\  E_{n,g}^{(2)} 
    = \frac {|\sqrt{n  } \hbar \Omega|^2 }
            {\hbar\left(
                  \omega_c n - {\omega_a \over 2} 
                - (\omega_c(n-1) + {\omega_a \over 2})
            \right)}
    = \frac {\hbar \Omega^2 n }
            {\omega_c - \omega_a}
\\  = {\hbar \Omega^2 \over \omega_a-\omega_c} - E_{n,e}.
$$
The term $\Delta = \omega_a - \omega_c$ is often called the detuning between the two systems, and it should be clear that detuning should be much greater than the coupling $|\Delta| \gg \Omega$ to keep the correction small in accordance with time-independent perturbation theory. In terms of detuning, the second-order approximation of eigenstate energies is thus

$$
\begin{align*}
      E_{n,e} 
   &=  \hbar \omega_c n 
    + {\hbar \omega_a   \over 2} 
    + {\hbar \Omega^2(n+1)
                        \over \Delta} 
\\ &=  \hbar \omega_c \left(n + {1\over 2} \right) 
    + {\hbar \Delta \over 2} 
    + {\hbar \Omega^2(n+1)
                        \over \Delta} 
\\ &=  \hbar \omega_c \left(n + {1\over 2} \right) 
    + {\hbar \over 2} {\left( 
        \Delta^2 + 2\Omega^2(n+1)
    \right) \over \Delta}
\\    E_{n-1,e} 
   &=  \hbar \omega_c \left(n - {1\over 2} \right) 
    + {\hbar \over 2} {\left( 
        \Delta^2 + 2\Omega^2n
    \right) \over \Delta}
\\    E_{n,g} 
   &=  \hbar \omega_c n
    - {\hbar \omega_a   \over 2} 
    - {\hbar \Omega^2 n \over \Delta}
\\ &=  \hbar \omega_c \left(n - {1\over 2} \right) 
    - {\hbar \over 2} {\left( 
        \Delta^2 + 2\Omega^2n
    \right) \over \Delta}
\end{align*}
$$

## Rabi Oscillations

In class, we discussed a variety of methods of introducing time dependence into a Hamiltonian, including time-dependent controlled pulses or continuous driving in the interaction representation and using [Fermi's second golden rule][2].

We cannot use Fermi's golden rule due to having entirely discrete states (no density of states), but we can use the other methods. Thanks to the nature of this Hamiltonian being block diagonal, Schrödinger's equation will describe natural oscillation between the two states. In addition, we can call $\Omega$ a perturbation from the product basis, or analyze $\Omega_0 + \delta\Omega(t)$ with the conserved total excitation basis.

Let's start without perturbation via Rabi oscillations, which come from the Schrödinger equation

$$
i\hbar {\partial \over \partial t} |\Psi \rangle = \hat H |\Psi \rangle
$$

To make use of this, we'll rotate into a diagonal conserved excitation basis (using $n$ as total excitation).
$$
\hat H_{\text{diag}} = 
{\hbar \over 2}
\begin{pmatrix}
(2n-1)\omega_c + \sqrt{\Delta^2 + 4\Omega^2 n} & 0 \\ 0 &
(2n-1)\omega_c - \sqrt{\Delta^2 + 4\Omega^2 n}
\end{pmatrix}
=
    R^T(\theta) \hat H R(\theta)
$$
Since a matrix is diagonalized by it's eigenvectors, we know
$$
    \ket{1} = \cos(\theta)\ket{n,g} + \sin(\theta)\ket{n,e}
\\  \ket{2} =-\sin(\theta)\ket{n,g} + \cos(\theta)\ket{n,e}
$$
Using Wolfram Mathematica to do the heavy lifting with the matrix multiplication, we have
$$
\Rightarrow
    \Delta \sin(2\theta) + 2\Omega \sqrt{n} \cos(2\theta) = 0
\\  \tan(2\theta) = -{2 \Omega \sqrt{n} \over \Delta}
\\  \theta = {1 \over 2} \arctan\left( - {2 \Omega \sqrt n \over \Delta}\right)
\\[1em]
$$
with the dress states
$$
    \sin(\arctan(x)) ={x \over \sqrt{x^2+1}}
\\  \cos(\arctan(x)) ={1 \over \sqrt{x^2+1}}
\def\Rn{\sqrt{\Delta^2+4\Omega^2n}}
\\  \ket{1} = \sqrt{{\Rn + \Delta \over 2\Rn}} \ket{n,e} + \sqrt{{\Rn - \Delta \over 2\Rn}} \ket{n,g}
\\  \ket{2} = \sqrt{{\Rn - \Delta \over 2\Rn}} \ket{n,e} + \sqrt{{\Rn - \Delta \over 2\Rn}} \ket{n,g}
$$
which isn't super pretty. In literature, the energy splitting $\mathcal R_n = \sqrt{\Delta^2+4\Omega^2n}$ is called the Rabi frequency, and in those terms the solution boils down to
$$
\def\Rn{\mathcal R_n}
\\  \ket{1} = \sqrt{{\Rn + \Delta \over 2\Rn}} \ket{n,e} + \sqrt{{\Rn - \Delta \over 2\Rn}} \ket{n,g}
\\  \ket{2} =-\sqrt{{\Rn - \Delta \over 2\Rn}} \ket{n,e} + \sqrt{{\Rn + \Delta \over 2\Rn}} \ket{n,g}
$$

## Time-Dependent Perturbation: Driven Dressed States

While the exact diagonalization solves the static case $\Omega_0$, we are often interested in driving transitions between these energy levels. Consider a system prepared in the lower dressed state $\ket{i} = \ket{2}$ subject to a small, time-dependent modulation of the coupling strength $\delta\Omega(t) = \epsilon \cos(\omega_d t)$.

We treat this modulation as a perturbation $V(t)$ acting on the diagonal basis. In the interaction representation, the first-order transition amplitude to the state $\ket{1}$ is given by

$$
\def\Rn{\mathcal R_n}

c_{1,2}^{(1)}(t) = \frac{1}{i\hbar} \int_0^t dt \braket{ 1 | \hat V(t) | 2 } e^{i(\omega_1 - \omega_2) t}.
$$

The resonant frequency is now the Rabi frequency $\Rn = \omega_+ - \omega_- = \sqrt{\Delta^2 + 4\Omega_0^2 n}$.

To evaluate the matrix element $\braket{ 1 | \hat V(t) | 2 }$, we must expand the interaction operator $\hat X = a^\dagger \sigma_- + a \sigma_+$ in the dressed basis. Using the angle $\theta$ derived previously:
$$
\braket{ 1 | \hat X | 2 } = \cos(2\theta)
$$
Substituting the perturbation $V(t) = \hbar \epsilon \cos(\omega_d t) \hat X$, the integral becomes:

$$
\begin{aligned}
c_{+-}^{(1)}(t) &= \frac{1}{i\hbar} \int_0^t dt \left[ \hbar \epsilon \cos(\omega_d t) \cos(2\theta) \right] e^{i \Rn t} \\
&= -i \epsilon \cos(2\theta) \int_0^t dt \frac{e^{i\omega_d t} + e^{-i\omega_d t}}{2} e^{i \Rn t}
\end{aligned}
$$

Since we assumed RWA earlier, we'll use $\omega_d \approx \Rn+\delta\Rn$ and assume any term $e^{i(\Rn + \omega_d)t}$ is thrown out. We keep only the slow difference term $\delta\Rn$:

$$
\begin{aligned}
c_{1,2}^{(1)}(t) &\approx \frac{-i \epsilon \cos(2\theta)}{2} \int_0^t e^{i \delta\Rn t} dt \\
&= -i \epsilon \cos(2\theta) e^{i \delta\Rn t / 2} \frac{\sin(\delta\Rn t / 2)}{\delta\Rn}
\\
P_{1}(t) &= |c_{1,2}^{(1)}(t)|^2
\\ &= {\epsilon^2 \cos^2(2\theta) \over \delta \Rn^2}\sin^2\left(\delta\Rn t \over 2 \right)
\end{aligned}
$$

This result indicates that we can drive coherent Rabi oscillations *between* the dressed states. The probability of finding the system in the state is $P_{1}(t) \propto \text{sinc}^2(\delta\Rn t/2)$, peaking when the drive frequency $\omega_d$ matches the dressed energy splitting.



## Adiabatic Evolution

Now consider the case where the coupling is not oscillating, but changing slowly (adiabatically) compared to the energy gap scale, $|\dot{\Omega}| \ll \Rn^2$. We choose a linear ramp function for the coupling, starting from zero:
$$
\def\Rn{\mathcal R_n}

\Omega(t) = \begin{cases} 
0 & t < 0 \\
\alpha t & t \geq 0 
\end{cases}
$$
In the adiabatic limit, the system remains in the instantaneous eigenstate of the Hamiltonian. If we initialize the system at $t=0$ in the state $\ket{\psi(0)} = \ket{2}$, the time-evolution is not described by transition probabilities $c_{fi}$, but by parameterized mixing angle $\theta(t)$.

Recall the definition of the mixing angle:
$$
\tan(2\theta(t)) = \frac{2\Omega(t)\sqrt{n}}{\Delta} = \frac{2\alpha t\sqrt{n}}{\Delta}
$$

At $t=0$, $\Omega(0)=0$ implies $\tan(2\theta) = 0$, so $\theta = 0$.
The dressed states at $t=0$ are:
$$
\begin{aligned}
\ket{1}_{t=0} &= \sin(0)\ket{n, g} + \cos(0)\ket{n, e} = \ket{n, e} \\
\ket{2}_{t=0} &= \cos(0)\ket{n, g} - \sin(0)\ket{n, e} = \ket{n, g} 
\end{aligned}
$$

As time evolves $t > 0$, the angle $\theta(t)$ grows. The state $\ket{\psi(t)}$ remains $\ket{1}$, but its physical character changes:
$$
\ket{\psi(t)} = \cos(\theta(t)) \ket{n, g} - \sin(\theta(t)) \ket{n, e}
$$
Notice the lack of Rabi Oscillations in the plots of adiabatic evolution. In class, we acknowledged that adiabatic evolutoin in the lab is never as perfect as the models, and states can still flip. Differential equation solvers can model the imperfections of adiabatic evolution, but the curves below take the idea of slow changes preventing state flips as certain. Also, all of this falls apart outside the RWA approximation, as some higher order effects can occur.
