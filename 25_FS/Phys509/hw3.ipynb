1. Commutation relation invariance under transformation from Schrödinger to Heisenberg picture.

Simple! We'll just expand out the Heisenberg version and show it becomes $\hat C_H(t)$
$$
\def\c#1#2{\left[#1, #2\right]}
\def\dag{\dagger}
    \c{\hat A_H(t)}{\hat B_H(t)} 
    = {\hat A_H(t)}{\hat B_H(t)} - {\hat B_H(t)}{\hat A_H(t)}
\\  = {U^\dag(t) \hat A_S U(t)}{U^\dag(t) \hat B_S U(t)} - {U^\dag(t) \hat B_S U(t)}{U^\dag(t) \hat A_S U(t)}
\\  = {U^\dag(t) \hat A_S \hat B_S U(t)} - {U^\dag(t) \hat B_S \hat A_S U(t)}
\\  = {U^\dag(t) (\hat A_S \hat B_S - \hat B_S \hat A_S) U(t)}
\\  = {U^\dag(t) \c{\hat A_S}{\hat B_S} U(t)}
\\  = {U^\dag(t) \hat C_S U(t)}
    = \hat C_H(t)
$$

2. Explicit forms of $ \hat{x}_H(t) $ and $ \hat{p}_H(t) $ using:
   a) Heisenberg’s equation of motion, and
   b) the unitary transformation $ \hat{O}_H(t) = e^{iHt/\hbar} \hat{O}_S e^{-iHt/\hbar} $ with the Baker–Hausdorff lemma.

$$
\hat H = \frac{\hat p^2}{2m} + \frac{1}{2} m \omega^2 \hat x^2
\\ = \hbar \omega (a^\dagger a + \frac{1}{2})
$$

a) The Heisenberg evolution equation is

$$
\frac{d}{dt} \hat A_H(t) = -\frac{1}{i\hbar} \c{\hat H}{\hat A_H(t)}
$$

So we just plug in $\hat x_H(t)$ and $\hat p_H(t)$

$$
\frac{d}{dt} \hat x_H(t) = -\frac{1}{i\hbar} \c{\hat H}{\hat x_H(t)}
\\ = -\frac{1}{i\hbar} \left[
    \frac{\hat p^2_H(t)}{2m} {\hat x_H(t)} + \frac{1}{2} m \omega^2 \hat x^2 {\hat x_H(t)}
   -{\hat x_H(t)}\frac{\hat p^2_H(t)}{2m} - \frac{1}{2} m \omega^2 \hat x^2{\hat x_H(t)}
   \right]
\\ = -\frac{1}{i\hbar}\frac{1}{2m} \left[
    \hat p^2_H(t)    \hat x_H(t)
   -\hat x_H(t) \hat p^2_H(t)
   \right]
$$
Hey! That looks familiar
$$
   \c{\hat p^2_H}{\hat x_H} = \hat p_H \c{\hat p_H}{\hat x_H} + \c{\hat p_H}{\hat x_H} \hat p_H
\\ = \hat p_H (-i\hbar) + (-i \hbar) \hat p_H   
\\ \frac{d}{dt} \hat x_H(t) = -\frac{1}{i\hbar}\frac{1}{2m} (-2i\hbar \hat p_H) = \frac{\hat p_H}{m}
$$

Similarly, we can drop the $\hat p^2$ term of the hamiltonian when we do $\frac{d}{dt} \hat p$

$$
\frac{d}{dt} \hat p_H = -\frac{1}{i\hbar} \c{\hat H}{\hat p_H} = -\frac{1}{i\hbar} \c{\frac{1}{2}m \omega^2 \hat x_H^2}{\hat p_H}
\\ = -\frac{m \omega^2}{2i\hbar}  \c{\hat x_H^2}{\hat p_H}
\\ = -\frac{m \omega^2}{2i\hbar}  (-2i\hbar x_H)
\\ = - m \omega^2 \hat x_H
$$

The $m$'s cancel out, so $\frac{d^2}{dt^2} \hat x_H = -\omega^2 x_H$ and we have

$$
   \hat x_H(0) = \hat x_S
\\ \hat p_H(0) = \hat p_S
\\ \hat x_H(t) = \hat x_S \cos( \omega t) + \frac{1}{\omega} \frac{\hat p_S}{m} \sin(\omega t)
\\ \hat p_H(t) =-m\omega \hat x_S \sin(\omega t) + \hat p_S \cos( \omega t) 
$$

b)

$$
    U(t) = e^{\frac{1}{i\hbar} \hat H t}
\\  \hat x_H(t) = e^{-\frac{1}{i\hbar} \hat H t} \hat x_S e^{\frac{1}{i\hbar} \hat H t}
$$

Apply Baker-Campbell-Hausdoff

$$
    \hat x_H(t) = \hat x_S - \frac{t}{i\hbar} \c{\hat H}{\hat x_S} - \frac{t^2}{\hbar^2 2!} \c{\hat H}{\c{\hat H}{\hat x_S}} + \dots
$$
Using the commutation relations we just found we can close the recusion formula 
$$
\\  \c{\hat H}{\hat x} = -\frac{i\hbar}{m} \hat p
\\  \c{\hat H}{\hat p} = i\hbar m \omega^2 \hat x
\\  C^{2n}  =  \hat x (\omega \hbar)^{2n}
\\  C^{2n+1}= -\hat p \frac{i}{m}(\omega \hbar)^{2n+1}
$$
Now recombine
$$
\\  \hat x_H(t) = \hat x \sum_{k=0}^{\infty} \frac{(-1)^n}{(2n)!}(\omega t)^{2n} - \frac{\hat p}{m\omega} \sum_{k=0}^\infty \frac{(-1)^n}{(2n+1)!}(\omega t)^{2n+1}
\\  \hat x_H(t) = \hat x \cos(\omega t) - \frac{\hat p}{m\omega} \sin(\omega t)
$$

5. Non-equal time commutators:
   - $[x_H(t), p_H(t')]$
   - $[a_H(t), a_H^\dagger(t')]$

   Now we have $U(t)$ and $U(t')$, which means we lose our proof from 1., which depended on equal time. Whatever our result is, we should end up with a solution that collapses to the above results for $t = t'$.

We can relate each term using our results from the previous problems

$$
   \hat x_H(t) = \hat x_S \cos(\omega t) + \frac{\hat p_S}{m\omega}\sin(\omega t)
\\ \hat p_H(t) = \hat p_S \cos(\omega t) - m\omega \hat x_S \sin(\omega t)
\\ [\hat x_S, \hat p_S] = i\hbar
$$

a)

Expand the commutator and drop vanishing terms ($[\hat x_S, \hat x_S] = [\hat p_S, \hat p_S] = 0$):

$$
[\hat x_H(t), \hat p_H(t')] = \left[
   \hat x_S \cos\omega t + \frac{\hat p_S}{m\omega}\sin\omega t,
   \hat p_S \cos\omega t' - m\omega \hat x_S \sin\omega t'
   \right]
\\ = \cos\omega t \cos\omega t' [\hat x_S, \hat p_S]
\times \frac{\sin\omega t}{m\omega}(-m\omega\sin\omega t') [\hat p_S, \hat x_S].
\\ = i\hbar\cos\omega t\cos\omega t' + i\hbar\sin\omega t\sin\omega t'
\\ = \boxed{i\hbar \cos\left(\omega (t - t')\right)}.
$$


b)

Using the time dependence from Problem 3:


$$
    \hat a_H(t) = \hat a_S e^{-i\omega t}
\\  \hat a_H^\dagger(t') = \hat a_S^\dagger e^{+i\omega t'}
\\  [\hat a_S, \hat a_S^\dagger] = 1
$$


$$
\begin{aligned}
[\hat a_H(t), \hat a_H^\dagger(t')]
    &= [\hat a_S e^{-i\omega t}, \hat a_S^\dagger e^{+i\omega t'}] 
\\  &= e^{-i\omega t} e^{+i\omega t'} [\hat a_S, \hat a_S^\dagger] 
\\  &= e^{-i\omega(t - t')}
\\  &= \boxed{e^{-i\omega(t - t')}}
\end{aligned}
$$
