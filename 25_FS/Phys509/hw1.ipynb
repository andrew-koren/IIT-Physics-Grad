Wouldn't it be nice if $C$ was diagonal? It would make this calculation so simple. Since we can unitary transfom matricies, let's work off diagonal ones and transform into arbitrary unitaries. 

1. The eigenvalues $\lambda_i$ of a unitary matrix have norm $|\lambda_i|^2=1$
2. A unitary matrix can be transformed into a diagonal unitary matrix with values equal to it's eigenvalues by unitary matricies.

A unitary can be transformed into the diagonal matrix $D$ by unitary matrix $Q$ made from it's eigenvectors.
$$
\\ D = Q^\dag U Q
\\ D_{ii} = \lambda_i 
$$
Since $\lambda_i$ have norm 1, they can be represented as a vector on a unit circle, so
$$
\\ \lambda_i = \cos(x) + i\sin(x) = e^{ix_i}
$$
where $x$ is the angle of the unit circle. 

Thus, we can make some real diagonal operator $X$ that, when exponentiated, yields the eigenvalues as it's rows.

$$
\\ X_{ii} = x_i \in \mathbb{R}
\\ X^n_{ii} = x_i^n
\\ e^{iX} = \1 + iX - \frac{X^2}{2} - i\frac{X^3}{6} + \dots
\\ (e^{iX})_{ii} = 1 + i X_{ii} - \frac{X^2_{ii}}{2} - i\frac{X^3_{ii}}{6} + \dots
\\= 1 + i x_i - \frac{x^2_{i}}{2} - i\frac{x^3_{i}}{6} + \dots
\\= e^{ix_i} = \lambda_i = D_{ii}
\\ D_{ij} =(e^{iX})_{ij}
\\ D = e^{iX}
$$


Thus, we can apply untary transform to get $U$

$$
U = Q^\dag D Q
\\ U = Q^\dag e^{iX} Q
\\ = Q^\dag \1 Q + Q^\dag iX Q - \frac{Q^\dag {X}^2 Q}{2} - \dots
$$

Each $X$ is simultaneously diagonalizable/transformable (obviously they commute), so if we write $C^n = Q^\dag {X}^n Q$, we have

$$
U = \1 + iC - \frac{C^2}{2} - i\frac{C^3}{6} + \dots = e^{iC}
\\ C =  Q^\dag {X}^n Q
$$

What about showing $C$ is hermitian? Since $X$ is real diagonal and $Q$ is unitary, we have

$$
C^\dag = \left(Q^\dag {X}^n Q \right)^\dag 
    = Q {X^n}^\dag Q^\dag
\\  = Q^\dag X^n Q = C
$$
