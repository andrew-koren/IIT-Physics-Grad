3. Baker-Hausdorff lemma

$$
\def\c#1#2{\left[ #1 , #2\right]}

e^{A}Be^{-A} = B 
    + \c{A}{B} 
    + \frac{1}{2!} \c{A}{\c{A}{B}}
    + \frac{1}{3!} \c{A}{\c{A}{\c{A}{B}}}
    + \dots
$$

The solution to this lemma requires we make use of Professor Shylnov's hint in class a few weeks back, to define a function $f(\lambda) = e^{\lambda A} B e^{-\lambda A}$ and take the taylor expansion with respect to $\lambda$

$$
\def\c#1#2{\left[#1,#2\right]}
\def\f#1{e^{\lambda A}#1 e^{-\lambda A}}

f(\lambda) = \f{B}

\\ \frac{d}{d\lambda} f(\lambda) = \f{AB} - \f{BA}
\\ = \f{\c{A}{B}}
$$

Apply some induction. Assumption:

$$
   \frac{d^k}{d\lambda^k} f(\lambda) 
    = \f{\underbrace{\c{A}{\c{A}{\dots \c{A}{B}}}}_{k \text{ times}}}
    
\\ \frac{d^{k+1}}{d\lambda^{k+1}} f(\lambda) 
    = \f{
         A\underbrace{\c{A}{\dots \c{A}{B}}}_{k \text{ times}}
        -\underbrace{\c{A}{\dots \c{A}{B}}}_{k \text{ times}} A
        }
\\  = \f{\underbrace{\c{A}{\c{A}{\dots \c{A}{B}}}}_{k+1 \text{ times}}}
