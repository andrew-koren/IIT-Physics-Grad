1. Use the second quantization formalism to calculate the first order non-linear correction to one-dimensional harmonic oscillator energy levels if the non-linear term is $\hat{\mathcal{H}}_I = \lambda {\hat p}^4 / 4$. 


$$
\\  \hat H = \hat H_0 + \hat H_I
\\  E_k = E_k^{(0)} + E_k^{(1)}
\\  E_n^{(0)} = \hbar \omega (n + 1/2)
\\  E_n^{(1)} = {}^{(0)}\!\braket{ k | \hat H_I | k }{}^{(0)} 
\\            = \braket{n | \lambda \hat p^4/4 | n}
$$
Using $\hat p = i\frac{\sqrt{m \hbar \omega}}{\sqrt 2}(\hat a^\dag - \hat a) $

$$
\\  \frac{\lambda}{4} \hat p^4 = {\lambda \over 16}(m \hbar \omega)^2
    \left( \hat a^\dag - \hat a \right)^4
\\  \braket{n | \lambda \hat p^4/4 | n} = {\lambda \over 16}(m \hbar \omega)^2 \braket{n|\left( \hat a^\dag - \hat a \right)^4|n}
$$

Let's look at that quartic. Mismatched $\hat a$ and $\hat a^\dag$ will have no overlap, so we only have matching terms. Considering only the matching terms, this expands to

$$
    \braket{n|(\hat a^\dag - \hat a)^4|n} 
    = \Braket{n|{\hat a^\dagger} \hat a {\hat a^\dagger} \hat a 
    + {\hat a^\dagger} \hat a^2 {\hat a^\dagger} 
    + \hat a {\hat a^\dagger} \hat a {\hat a^\dagger} 
    + \hat a^2 {\hat a^\dagger}^2 |n}
\\  = \Braket{n|n} \left[
      n ^2 
    + n(n +1) 
    + (n + 1)^2  
    + (n + 1)(n + 2) 
    \right]
\\  E_n^{(1)} = \frac{\lambda}{16} (m\hbar \omega )^2 \left(4n^2 + 6n + 3 \right)
$$

3. Calculate energy levels of a two-dimensional isotropic oscillator with the Hamiltonian of 

$$ \hat H(𝑥,𝑦)= \frac{1}{2𝑚} (𝑝̂_𝑥^2 + 𝑝̂_𝑦^2) +\frac{1}{2}𝑘 (𝑥̂^2 +𝑦̂^2) +𝛼𝑥̂𝑦̂ 
$$
Consider the case of $|𝛼|<𝑘$. What happens if $|𝛼|>𝑘$ ? 
Hints: Diagonalize the potential energy quadratic form. Double check that the Laplace operator from the kinetic energy term is still diagonal in new coordinates. Now, you have a sum of two independent one-dimensional harmonic oscillators. 


The potential energy term is $\hat V=\frac{1}{2} k (\hat x^2+ \hat y^2) + \alpha \hat x \hat y$. To diagonalize, we'll convert this to a matrix and find the eigenvalues/vectors

$$
\hat V = \frac{1}{2}
    \begin{pmatrix} \hat x & \hat y \end{pmatrix}
    \begin{pmatrix} k      & \alpha \\
                    \alpha & k      \end{pmatrix}
    \begin{pmatrix} \hat x\\ \hat y \end{pmatrix}
$$
This works because $[x,y]=0$. 
$$
    (\hat V - \lambda I) = 0
\\  (k-\lambda)^2 - \alpha^2 = 0
\\  \lambda = k \pm \alpha
\\  \begin{pmatrix} k & \alpha \\
                    \alpha & k \end{pmatrix}
    \begin{pmatrix} 1 \\ 1 \end{pmatrix} 
    = (k+\alpha) \begin{pmatrix} 1 \\ 1 \end{pmatrix} 
\\  \begin{pmatrix} k & \alpha \\
                    \alpha & k \end{pmatrix}
    \begin{pmatrix} 1 \\ -1 \end{pmatrix} 
    = (k-\alpha) \begin{pmatrix} 1 \\ -1 \end{pmatrix} 
\\  \hat u = {1 \over \sqrt{2}} \left(\hat x + \hat y\right)
\\  \hat v = {1 \over \sqrt{2}} \left(\hat x - \hat y\right)
$$
With the square root for normalization (keeps diagonalization consistent). This gives us
$$
\hat V = \frac{1}{2}
    \begin{pmatrix} \hat u &  \hat v\end{pmatrix}
    \begin{pmatrix} k + \alpha & 0 \\ 0 & k-\alpha  \end{pmatrix} 
    \begin{pmatrix} \hat u \\ \hat v\end{pmatrix}
\\  = \frac{1}{2} (k + \alpha)\hat u^2 + \frac{1}{2} (k-\alpha) \hat v^2
$$
It looks like kinetic energy works out fine too, since the laplace operator is unchanged by rotation and the change of basis is the same as a $\pi/4$ rotation.
$$
\frac{1}{2m} (\hat p_x^2 + \hat p_y^2) = \frac{1}{2m} (\hat p_u^2 + \hat p_v^2)
$$
The separated Hamiltonian is thus
$$
\hat H(x,y) = H_u(u) + H_v(v)
    = \frac{1}{2} \left(\frac{\hat p_u^2}{m} + (k + \alpha)\hat u^2 \right)
    + \frac{1}{2} \left(\frac{\hat p_u^2}{m} + (k - \alpha)\hat u^2 \right)

$$

Now we can apply our general solution to the 1D harmonic oscillator
$$
k \pm \alpha = m \omega_{\pm}^2
\\ \omega_{\pm} = \sqrt{\frac{k \pm \alpha}{m}}
$$
We focus on $k > |\alpha|$ so that the system remains stable. If $\omega$ is complex, the wavefunction will not oscillate and will instead grow exponentially.

$$
E = E_u + E_v = 
      \hbar \omega_{+} \left(n_u + \frac{1}{2} \right)
    + \hbar \omega_{-} \left(n_v + \frac{1}{2} \right)
$$

Where $n_u = \braket{\hat a_u^\dag \hat a_u}$, etc. following the typical definition of the creation/annihilation operators only with $u,v$ and $\omega_{\pm}$ instead of $x,y$ and $\omega$. Not so isotropic now, is it?
