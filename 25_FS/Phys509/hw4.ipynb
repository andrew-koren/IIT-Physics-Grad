3. The potential energy term is $\hat V=\frac{1}{2} k (\hat x^2+ \hat y^2) + \alpha \hat x \hat y$. To diagonalize, we'll convert this to a matrix and find the eigenvalues/vectors

$$
\hat V = \frac{1}{2}
    \begin{pmatrix} \hat x & \hat y \end{pmatrix}
    \begin{pmatrix} k      & \alpha \\
                    \alpha & k      \end{pmatrix}
    \begin{pmatrix} \hat x\\ \hat y \end{pmatrix}
$$
This works because $[x,y]=0$. 
$$
    (\hat V - \lambda I) = 0
\\  (k-\lambda)^2 - \alpha^2 = 0
\\  \lambda = k \pm \alpha
\\  \begin{pmatrix} k & \alpha \\
                    \alpha & k \end{pmatrix}
    \begin{pmatrix} 1 \\ 1 \end{pmatrix} 
    = (k+\alpha) \begin{pmatrix} 1 \\ 1 \end{pmatrix} 
\\  \begin{pmatrix} k & \alpha \\
                    \alpha & k \end{pmatrix}
    \begin{pmatrix} 1 \\ -1 \end{pmatrix} 
    = (k-\alpha) \begin{pmatrix} 1 \\ -1 \end{pmatrix} 
\\  \hat u = {1 \over \sqrt{2}} \left(\hat x + \hat y\right)
\\  \hat v = {1 \over \sqrt{2}} \left(\hat x - \hat y\right)
$$
With the square root for normalization (keeps diagonalization consistent). This gives us
$$
\hat V = \frac{1}{2}
    \begin{pmatrix} \hat u &  \hat v\end{pmatrix}
    \begin{pmatrix} k + \alpha & 0 \\ 0 & k-\alpha  \end{pmatrix} 
    \begin{pmatrix} \hat u \\ \hat v\end{pmatrix}
\\  = \frac{1}{2} (k + \alpha)\hat u^2 + \frac{1}{2} (k-\alpha) \hat v^2
$$
It looks like kinetic energy works out fine too, since the laplace operator is unchanged by rotation and the change of basis is the same as a $\pi/4$ rotation.



$$
\frac{1}{2m} (\hat p_x^2 + \hat p_y^2) = \frac{1}{2m} (\hat p_u^2 + \hat p_v^2)
$$
The separated Hamiltonian is thus
$$
\hat H(x,y) = H_u(u) + H_v(v)
    = \frac{1}{2} \left(\frac{\hat p_u^2}{m} + (k + \alpha)\hat u^2 \right)
    + \frac{1}{2} \left(\frac{\hat p_u^2}{m} + (k - \alpha)\hat u^2 \right)

$$

Now we can apply our general solution to the 1D harmonic oscillator
$$
k \pm \alpha = m \omega_{\pm}^2
\\ \omega_{\pm} = \sqrt{\frac{k \pm \alpha}{m}}
$$
We focus on $k > |\alpha|$ so that the system remains stable. If $\omega$ is complex, the wavefunction will not oscillate and will instead grow exponentially.

$$
E = E_u + E_v = 
      \hbar \omega_{+} \left(n_u + \frac{1}{2} \right)
    + \hbar \omega_{-} \left(n_v + \frac{1}{2} \right)
$$

Where $n_u = \braket{\hat a_u^\dag \hat a_u}$, etc. following the typical definition of the creation/annihilation operators only with $u,v$ and $\omega_{\pm}$ instead of $x,y$ and $\omega$. Not so isotropic now, is it?

6. Using interaction representation and $\hat V(t) = q \mathcal E(t) \hat x$

$$
    V_I(t) = e^{-H_0 t \over i \hbar} V(t) e^{H_0 t \over i \hbar}
\\  \ket{n, t, t_0} = U_I(t,t_0)\ket{n}
\\  U_I(\infty, -\infty) = Te^{1\over i \hbar} \int_{-\infty}^\infty V_I(t) dt
\\  \hat H_0 = \hbar \omega\left(\hat N + \frac{1}{2}\right)
$$
Using $\ket n = \ket{n_i}$ and $\ket m = \ket{n_f}$, we have the transition amplitude $\braket{m|U_I|n} $ as

$$
\braket{m|U_I|n} = \delta_{mn} + {1 \over i \hbar} \int_{-\infty}^\infty e^{i \omega_{mn}t} V_{mn}(t) dt
$$
To evaluate this, we'll need to find $V_{mn}(t)$, so convert $\hat x$ to $(\alpha+\alpha^\dag)$

Using interaction representation and Stark effect $\hat V(t) = -e \mathcal E(t) \hat x$, the time-dependent perturbation is


$$
\braket{m|U_I|n} = \delta_{mn} + {1 \over i \hbar} \int_{-\infty}^\infty e^{i \omega_{mn}t} V_{mn}(t) dt

\\  \hat x 
    = \left( \hbar \over 2 m \omega \right)^{1/2} \left(a^\dag + a\right) 
    = a_0\left(a^\dag + a\right)
\\  V_{mn} 
    =q a_0 \mathcal{E}(t) \braket{m | a^\dag + a | n} 
    =q a_0 \mathcal{E}(t) \left[\sqrt{n+1} \delta_{m,n+1} 
                                + \sqrt{n}   \delta_{m, n-1} \right]
$$

So, it is clear that the pulse has nonzero first-order effects from the $n\text{th}$ state to the $n\pm 1\text{th}$ state. This does mean $|\braket{n|U_I|n}|^2 = 1$, but I think that is expected in perturbation theory, since the other amplitudes should be small.

$$
      \braket{n\pm 1|U_I|n} 
    = {1 \over i \hbar} \int_{-\infty}^\infty q a_0 \sqrt{n+\frac{1\pm 1}{2}} e^{\pm i \omega t} \mathcal{E}(t)  dt
\\  = {1 \over i \hbar} q \mathcal{E}_0a_0 \sqrt{n+\frac{1\pm 1}{2}} 
      \int_{-\infty}^\infty e^{-t^2/\tau^2 \pm i \omega t} \cos(\omega_0 t) dt
$$
Looks like a bad integral, but we can use $\cos(\omega_0 t) = \frac{e^{i\omega_0 t} + e^{-i \omega_0 t}}{2}$

$$
\\  = {q \mathcal{E}_0a_0 \over 2 i \hbar}  \sqrt{n+\frac{1\pm 1}{2}} 
      \int_{-\infty}^\infty \left[
          e^{-t^2/\tau^2 + i (\omega_0 \pm \omega) t} 
        +  e^{-t^2/\tau^2 + i (-\omega_0 \pm \omega) t} 
        \right]dt

$$



```python
from sympy import symbols, integrate, I, oo, exp, simplify, conjugate
from sympy.abc import t

tau, omega, omega0 = symbols('tau, omega, omega0', real=True, positive=True)

# m = n+1
pos = - (t*t/(tau*tau)) + I*( omega0+omega)*t
neg = - (t*t/(tau*tau)) + I*(-omega0+omega)*t

result = simplify(integrate(exp(pos)+exp(neg), (t, -oo, oo)))
display(result)

# m = n-1 is the same
pos = - (t*t/(tau*tau)) + I*( omega0-omega)*t
neg = - (t*t/(tau*tau)) + I*(-omega0-omega)*t

result = simplify(integrate(exp(pos)+exp(neg), (t, -oo, oo)))
display(result)
```

I'm just glad there's no error function. This gives the result for probability amplitude:

$$
      \braket{n \pm 1 | U_I | n} 
    = {q \mathcal{E}_0a_0\tau \sqrt{\pi}\over 2 i \hbar}  \sqrt{n+\frac{1\pm 1}{2}} \left[
         e^{- \dfrac{\tau^{2} \left(\omega + \omega_{0}\right)^{2}}{4}} + e^{- \dfrac{\tau^{2} \left(\omega - \omega_{0}\right)^{2}}{4}}
    \right]
$$
That's a lot of constants, but the $n\pm 1$ term only affects if there is a $\sqrt{n}$ or $\sqrt{n+1}$ term.

```python
const = symbols('a', real=True)
amplitude = const/I * result
p = amplitude*conjugate(amplitude)

display(simplify((p)))
display(simplify(expand(p)))
```

The above expression is the probability of transitioning, where all the constants out front in the amplitude are collected into $a$
