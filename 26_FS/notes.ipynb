Quantum average

$$
\def\Tr{\text{Tr}}
\def\d{\partial}

        \braket{\hat A } 
    =   \sum_\alpha p_\alpha \braket{\psi_\alpha | \hat A | \psi_\alpha}
\\  =   \sum_{\alpha,m,n} p_\alpha \braket{\psi_\alpha | m} \braket{m | \hat A | n} \braket{n | \psi_\alpha}
\\  =   \sum_{m,n} \hat p \braket{m | \hat A | n} % density operator
$$
We use typical quantum matrix element notation, Tr is a single sum, etc

$$
        \braket{A} = \Tr \rho \hat A
\\      \rho 
    =   {e^{-\beta \left( \hat H - \mu \hat N\right)} \over \mathcal{L}}      
\\      \mathcal{L}
    =   \Tr e^{-\beta \left( \hat H - \mu \hat N \right)}
    =   \Tr \sum_{N=0}^\infty e^{\beta \mu N} Z_N
$$
where $Z_N$ comes from the canonical ensemble $Z = \Tr e^{-\beta \hat H}$

Derive Fermi-Dirac. $\epsilon_k$ is the energy of state $k$, while $N_k$ is total $k$-state particles. Fermions $N_k \in \{0,1\}$.

T in energy

$$
        \Omega
    = - T \ln \sum_{N_k}e^{\beta \mu N_k}e^{-\epsilon_k \beta N_k}
\\  = - T \ln \sum_k \left[e^{-\beta(\epsilon_k - \mu)}\right]^{N_k}
\\  = -T\ln\left\{ 1 + e^{-\beta(\epsilon_k - \mu )} \right\}
$$

Using $\braket{N_k} = -{\d \Omega \over \d \mu} = {1 \over 1 + e^{\beta(\epsilon_k - \mu)}}$
