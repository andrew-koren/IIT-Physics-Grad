1. show ${\kappa_S \over \kappa_T} = {C_V \over C_p} $ 

$$
\def\d{\partial}
\begin{gathered}
\kappa_S = -{1 \over V} \left(\d V \over \d p \right)_S,
& \quad &
\kappa_T = -{1 \over V} \left( \d V \over \d p \right)_T,
\\
C_V = T \left(\d S \over \d T \right)_V,
& \quad &
C_p = T \left( \d S \over \d T \right)_p.
\end{gathered}
$$
so
$$
      {\kappa_S \over \kappa_T} 
    = \frac{\left(\d V \over \d p \right)_S}
           {\left(\d V \over \d p \right)_T}.
$$

We need some $\d S / \d T$'s to to get $C_V$ and $C_p$, so we'll use the triple product rule to excrete $\d S$ and $\d T$:

$$
\begin{gathered}
        \left(\d V \over \d p \right)_S 
    = - \frac{\left(\d S \over \d p \right)_V}
             {\left(\d S \over \d V \right)_p}
    % = - {C_V \over T}
    %     \frac{1}
    %          {\left(\d S \over \d V \right)_p}
\\  
        \left(\d V \over \d p \right)_T 
    = - \frac{\left(\d T \over \d p \right)_V}
             {\left(\d T \over \d V \right)_p}
    % = - \frac{C_p}
    %          {\left(\d S \over \d V \right)_p}
\\      {\kappa_S \over \kappa_T} 
    =   \frac{\left(\d S \over \d p \right)_V}{\left(\d T \over \d p \right)_V}
        \frac{\left(\d T \over \d V \right)_p}{\left(\d S \over \d V \right)_p}
    =   {\left(\d S \over \d T \right)_V}
        \frac{1}{\left(\d S \over \d T \right)_p} = {T C_V \over T C_p} = {C_V \over C_p}
        
   
\end{gathered}
$$
